# 1. Strings

A **string** is text written inside quotes.

You can use:
- single quotes: `'Hello'`
- double quotes: `"Hello"`

A string can contain letters, numbers, spaces, and symbols.

### Example

````python
name = "Saad"
city = "Karachi"

print(name)
print(city)
````

### Line-by-line explanation

**Line 1**
```python
name = "Saad"
```
Creates a variable named `name` and stores the text `"Saad"` in it.

**Line 2**
```python
city = "Karachi"
```
Creates another variable and stores `"Karachi"`.

**Line 4**
```python
print(name)
```
Displays the value stored in `name`.

**Line 5**
```python
print(city)
```
Displays the value stored in `city`.

# 2. Getting Input from the User

`input()` lets your program ask the user for information.

The important thing to remember is:

> **`input()` returns text (a string).**

Even if the user types `18`, Python initially receives `"18"` as text.

````python
name = input("What is your name? ")

print("Hello,", name)
````

### Line-by-line explanation

**Line 1**
```python
name = input("What is your name? ")
```
Python displays the question and waits for the user to type something.

Whatever the user types is stored in `name` as a **string**.

**Line 3**
```python
print("Hello,", name)
```
Prints `"Hello,"` and then the user's name.

# 3. Type Conversion

Sometimes input is text, but we need a number.

Python provides functions such as:

```python
int()      # convert to an integer
float()    # convert to a decimal number
str()      # convert to a string
```

This is called **type conversion** (or type casting).

````python
age_text = input("Enter your age: ")
age = int(age_text)

print(age)
print(type(age))
````

### Line-by-line explanation

**Line 1**
```python
age_text = input("Enter your age: ")
```
Gets the user's input. Because `input()` returns text, `age_text` is a string.

**Line 2**
```python
age = int(age_text)
```
Converts that text into an integer and stores it in `age`.

**Line 4**
```python
print(age)
```
Displays the number.

**Line 5**
```python
print(type(age))
```
Shows Python's data type for `age`. Because we used `int()`, it is an integer.

### Important beginner mistake

This:
```python
age = input("Age: ")
```
does **not** make `age` an integer.

If you need a number:
```python
age = int(input("Age: "))
```



*input() always returns text — convert it with int() or float() before doing math.*

# 4. Arithmetic Operators

Arithmetic operators allow Python to perform calculations.

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `//` | Floor division |
| `%` | Remainder |
| `**` | Exponent |

### Example

````python
a = 17
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
````

### Line-by-line explanation

**Lines 1–2** create two numbers.

```python
a + b
```
adds them.

```python
a - b
```
subtracts `b` from `a`.

```python
a * b
```
multiplies them.

```python
a / b
```
performs normal division and returns a decimal result when necessary.

```python
a // b
```
performs floor division.

```python
a % b
```
returns the remainder.

```python
a ** 2
```
raises `a` to the power of 2.

# 5. Comparison Operators

Comparison operators compare values and produce a Boolean result:

- `True`
- `False`

Common comparison operators are:
`==`, `!=`, `>`, `<`, `>=`, `<=`.

````python
age = 18

print(age == 18)
print(age != 20)
print(age > 16)
print(age < 18)
print(age >= 18)
print(age <= 18)
````

### Line-by-line explanation

```python
age == 18
```
Checks whether `age` is equal to `18`.

```python
age != 20
```
Checks whether `age` is not equal to `20`.

```python
age > 16
```
Checks whether `age` is greater than `16`.

```python
age < 18
```
Checks whether `age` is less than `18`.

```python
age >= 18
```
Checks whether `age` is greater than or equal to `18`.

```python
age <= 18
```
Checks whether `age` is less than or equal to `18`.

### Important
`=` means **assignment**.

`==` means **comparison**.

# 6. Boolean Values and Logical Operators

A Boolean has only two possible values:

```python
True
False
```

Logical operators combine or change Boolean conditions:

- `and` → both conditions must be true
- `or` → at least one condition must be true
- `not` → reverses the Boolean result

````python
age = 18
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
````

### Line-by-line explanation

```python
age = 18
```
Stores the age.

```python
has_id = True
```
Stores a Boolean value.

```python
age >= 18 and has_id
```
Both conditions must be true. Here they are, so the result is `True`.

```python
age < 18 or has_id
```
At least one condition must be true. `has_id` is true, so the result is `True`.

```python
not has_id
```
Reverses `True` to `False`.
