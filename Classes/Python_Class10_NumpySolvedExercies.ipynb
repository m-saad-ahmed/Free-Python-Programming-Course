### Exercise 27 — Consider an integer vector Z, which of these expressions are legal? (★★☆)

```python
Z = np.arange(10)

# We try each expression in its own try/except (Class 8 style) so one
# illegal expression doesn't stop us from seeing the others.
expressions = {
    "Z**Z": lambda: Z**Z,
    "2 << Z >> 2": lambda: 2 << Z >> 2,
    "Z <- Z": lambda: Z <- Z,   # parsed as Z < (-Z)
    "1j*Z": lambda: 1j*Z,
    "Z/1/1": lambda: Z/1/1,
    "Z<Z>Z": lambda: Z<Z>Z,
}

for label, expr in expressions.items():
    try:
        expr()
        print(f"{label:12s} -> legal")
    except Exception as e:
        print(f"{label:12s} -> ILLEGAL ({type(e).__name__}: {e})")
```

**💡 Key idea:** Surprisingly, only the last expression is illegal: chained comparisons like `Z<Z>Z` try to convert a whole array's comparison into a single `True`/`False`, which NumPy refuses to guess (the same ambiguity error you'd get writing `if some_array:`).

### Exercise 52 — Consider a random vector with shape (100,2) representing coordinates, find point by point distances (★★☆)

```python
Z = np.random.random((10,2))
X,Y = np.atleast_2d(Z[:,0], Z[:,1])
D = np.sqrt( (X-X.T)**2 + (Y-Y.T)**2)
print(D)

# Much faster with scipy
import scipy
# Thanks Gavin Heverly-Coulson (#issue 1)
import scipy.spatial

Z = np.random.random((10,2))
D = scipy.spatial.distance.cdist(Z,Z)
print(D)
```

**💡 Key idea:** Computing every pairwise distance between 2D points by hand uses broadcasting twice; `scipy.spatial.distance.cdist()` does the same thing as one optimized library call.

### Exercise 56 — Generate a generic 2D Gaussian-like array (★★☆)

```python
X, Y = np.meshgrid(np.linspace(-1,1,10), np.linspace(-1,1,10))
D = np.sqrt(X*X+Y*Y)
sigma, mu = 1.0, 0.0
G = np.exp(-( (D-mu)**2 / ( 2.0 * sigma**2 ) ) )
print(G)
```

**💡 Key idea:** Combining a distance grid (`np.sqrt(X*X+Y*Y)`) with `np.exp()` produces the classic bell-curve shape of a 2D Gaussian.

### Exercise 78 — Consider 2 sets of points P0,P1 describing lines (2d) and a point p, how to compute distance from p to each line i (P0[i],P1[i])? (★★★)

```python
def distance(P0, P1, p):
    T = P1 - P0
    L = (T**2).sum(axis=1)
    U = -((P0[:,0]-p[...,0])*T[:,0] + (P0[:,1]-p[...,1])*T[:,1]) / L
    U = U.reshape(len(U),1)
    D = P0 + U*T - p
    return np.sqrt((D**2).sum(axis=1))

P0 = np.random.uniform(-10,10,(10,2))
P1 = np.random.uniform(-10,10,(10,2))
p  = np.random.uniform(-10,10,( 1,2))
print(distance(P0, P1, p))
```

**💡 Key idea:** Point-to-line-segment distance is a classic geometry formula, implemented here with vectorized NumPy operations instead of a manual loop over each line.

### Exercise 80 — Consider an arbitrary array, write a function that extract a subpart with a fixed shape and centered on a given element (pad with a fill value when necessary) (★★★)

```python
import numpy as np

def extract(Z, shape, position, fill=0):
    R = np.ones(shape, dtype=Z.dtype) * fill

    Z_h, Z_w = Z.shape
    R_h, R_w = shape
    y, x = position

    y1 = max(0, y - R_h // 2)
    y2 = min(Z_h, y + R_h // 2 + 1)
    x1 = max(0, x - R_w // 2)
    x2 = min(Z_w, x + R_w // 2 + 1)

    Ry1 = max(0, R_h // 2 - y)
    Rx1 = max(0, R_w // 2 - x)

    R[Ry1:Ry1 + (y2 - y1), Rx1:Rx1 + (x2 - x1)] = Z[y1:y2, x1:x2]
    return R
```

**💡 Key idea:** This function safely extracts a sub-window around any position, using `max()`/`min()` to clip the window at the array's edges and padding the rest with a fill value.

### Exercise 88 — How to implement the Game of Life using numpy arrays? (★★★)

```python
# Author: Nicolas Rougier

def iterate(Z):
    # Count neighbours
    N = (Z[0:-2,0:-2] + Z[0:-2,1:-1] + Z[0:-2,2:] +
         Z[1:-1,0:-2]                + Z[1:-1,2:] +
         Z[2:  ,0:-2] + Z[2:  ,1:-1] + Z[2:  ,2:])

    # Apply rules
    birth = (N==3) & (Z[1:-1,1:-1]==0)
    survive = ((N==2) | (N==3)) & (Z[1:-1,1:-1]==1)
    Z[...] = 0
    Z[1:-1,1:-1][birth | survive] = 1
    return Z

Z = np.random.randint(0,2,(50,50))
for i in range(100): Z = iterate(Z)
print(Z)
```

**💡 Key idea:** Conway's Game of Life counts each cell's living neighbors with shifted array slices, then applies the birth/survival rules to the whole grid at once — no per-cell loop needed.

```python
# Author: Alex Riley
# Make sure to read: http://ajcr.net/Basic-guide-to-einsum/

A = np.random.uniform(0,1,10)
B = np.random.uniform(0,1,10)

np.einsum('i->', A)       # np.sum(A)
np.einsum('i,i->i', A, B) # A * B
np.einsum('i,i', A, B)    # np.inner(A, B)
np.einsum('i,j->ij', A, B)    # np.outer(A, B)
```

**💡 Key idea:** `np.einsum()` is a single, flexible notation that can express sum, element-wise multiply, inner product, and outer product just by changing its subscript pattern.

### Exercise 98 — Considering a path described by two vectors (X,Y), how to sample it using equidistant samples ? (★★★)

```python
# Author: Bas Swinckels

phi = np.arange(0, 10*np.pi, 0.1)
a = 1
x = a*phi*np.cos(phi)
y = a*phi*np.sin(phi)

dr = (np.diff(x)**2 + np.diff(y)**2)**.5 # segment lengths
r = np.zeros_like(x)
r[1:] = np.cumsum(dr)                # integrate path
r_int = np.linspace(0, r.max(), 200) # regular spaced path
x_int = np.interp(r_int, r, x)       # integrate path
y_int = np.interp(r_int, r, y)
```

**💡 Key idea:** By measuring cumulative distance traveled along a path (`np.cumsum`) and using `np.interp()` to resample it, this produces evenly-spaced points along an irregular curve.

### Exercise 99 — Given an integer n and a 2D array X, select from X the rows which can be interpreted as draws from a multinomial distribution with n degrees, i.e., the rows which only contain integers and which sum to n. (★★★)

```python
# Author: Evgeni Burovski

X = np.asarray([[1.0, 0.0, 3.0, 8.0],
                [2.0, 0.0, 1.0, 1.0],
                [1.5, 2.5, 1.0, 0.0]])
n = 4
M = np.logical_and.reduce(np.mod(X, 1) == 0, axis=-1)
M &= (X.sum(axis=-1) == n)
print(X[M])
```

**💡 Key idea:** `np.mod(X, 1) == 0` checks that every value in a row is a whole number, and combined with a row-sum check, filters out rows that could plausibly be counts from `n` trials.

### Exercise 100 — Compute bootstrapped 95% confidence intervals for the mean of a 1D array X (i.e., resample the elements of an array with replacement N times, compute the mean of each sample, and then compute percentiles over the means). (★★★)

```python
# Author: Jessica B. Hamrick

X = np.random.randn(100) # random 1D array
N = 1000 # number of bootstrap samples
idx = np.random.randint(0, X.size, (N, X.size))
means = X[idx].mean(axis=1)
confint = np.percentile(means, [2.5, 97.5])
print(confint)
```

**💡 Key idea:** **Bootstrapping** estimates uncertainty by resampling the data (with replacement) thousands of times and looking at the spread of the resulting means — a widely used statistics technique.
