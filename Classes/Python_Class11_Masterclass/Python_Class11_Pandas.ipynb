# Module 1 — What Is Pandas?

Pandas is an open-source Python library for data analysis and manipulation. Its central tabular structure is the **DataFrame**, while a **Series** is a one-dimensional labeled structure.

### Mental model

Think:

**DataFrame = table + labels + programmable operations**

Excel is a useful analogy, but Pandas is not simply “Excel in Python”. Pandas gives us programmable, repeatable operations over labeled data.

### The core workflow

```text
SOURCE
  ↓
READ
  ↓
INSPECT
  ↓
CLEAN
  ↓
SELECT / FILTER
  ↓
TRANSFORM
  ↓
GROUP / AGGREGATE
  ↓
COMBINE / RESHAPE
  ↓
VALIDATE
  ↓
EXPORT
```

A professional analyst should be able to explain every transition in this pipeline.

## 🔎 Detailed Explanation — Module 1

Pandas is a library for working with structured, labeled data. The most important idea is that Pandas gives us a table-oriented way to represent data while keeping row and column labels attached to the values. A DataFrame is the main two-dimensional structure; a Series is a single labeled column.

Think of a DataFrame as a spreadsheet that Python can reason about programmatically. Each column has a name and a data type, each row has an index label, and Pandas operations can transform the whole table without requiring you to manually loop through every cell. This mental model is more important than memorizing commands.

A typical Pandas workflow is: load data → inspect it → select what matters → clean it → transform it → analyze it → validate the result. Each step answers a different question, and skipping inspection or validation is a common source of incorrect analysis.

### 🎯 What you should be able to explain

Understand the table before writing transformations. Ask: What does one row represent? What does each column mean? Which column identifies an observation? What values are missing or inconsistent?

# Module 2 — Series Deep Dive

A Series has:
- values
- an index
- a name
- a dtype

```text
index      values
  0   →      90
  1   →      75
  2   →      88
```

The index is a label, not necessarily a row number.

## 🔎 Detailed Explanation — Module 2

A Series is a one-dimensional labeled array. It contains values plus an Index, and it can also have a name and a dtype. The index is not merely decoration: it is used for selection and alignment. Two Series can align by labels even when their row positions differ.

You can create a Series from a list, dictionary, or scalar. A list normally receives a default integer index, while a dictionary naturally supplies labels from its keys. Selecting one DataFrame column returns a Series, which is why learning Series behavior is essential for almost every Pandas task.

Distinguish positional thinking from label thinking. `iloc` is position-based; `loc` is label-based. Also remember that vectorized operations act on the entire Series and normally preserve its index.

# Module 3 — DataFrame Deep Dive

A DataFrame is a two-dimensional labeled data structure.

```text
             columns
          Name  Age  City
index  0   Ali  20   KHI

```

A DataFrame has:
- an index
- column labels
- values
- dtypes
- shape

## 🔎 Detailed Explanation — Module 3

A DataFrame is a two-dimensional labeled data structure made from aligned columns. Every column is a Series sharing the same row index. This means a DataFrame is not just a list of lists: it has a schema-like structure where column names and data types matter.

When you select one column, you usually get a Series. When you select multiple columns, you get another DataFrame. Rows can be selected by labels with `loc` or by integer positions with `iloc`. This distinction becomes increasingly important when indexes are customized.

A useful habit is to inspect `shape`, `columns`, `index`, and `dtypes` before making assumptions. When you add a column, Pandas aligns values by index, so understanding labels prevents subtle errors.

### 🎯 What you should be able to explain

Think of a DataFrame as aligned Series objects sharing one index. Every row should have a clear meaning, and every column should have a clear meaning and expected type.

## Creating DataFrames safely

# Module 4 — Index and Labels

The index identifies rows through labels.

Default:

```text
0
1
2
```

But labels can be strings, dates, IDs, or other appropriate labels.

**Key distinction:**

- `loc` → label-based selection
- `iloc` → position-based selection

## 🔎 Detailed Explanation — Module 4

An Index provides labels for rows and, in some operations, columns. The default index is usually 0, 1, 2, …, but an index can contain strings, dates, identifiers, or multiple levels. Labels support selection, alignment, and time-based workflows.

The key distinction is `loc` versus `iloc`: `loc` asks for labels, while `iloc` asks for integer positions. Mixing these mental models is one of the easiest ways to select the wrong data. Pandas also uses index alignment during arithmetic and assignment, which means matching labels can matter more than matching positions.

Resetting an index does not simply “delete” information; it changes how row labels are represented. Setting an index can make certain operations clearer, especially when the chosen label has real meaning.

### 🎯 What you should be able to explain

Always ask whether the task is label-based or position-based. If you cannot state the row labels clearly, inspect `df.index` before selecting.

# 👀 Visual: `loc` vs `iloc`

```text
DataFrame index labels:
┌───────┬────────┐
│ Label │  Name  │
├───────┼────────┤
│  101  │  Ali   │
│  205  │ Ahmed  │
│  500  │  Saad  │
└───────┴────────┘

loc  → "Give me label 205"
       ↓
      Ahmed

iloc → "Give me position 1"
       ↓
      Ahmed
```

The result happens to be the same here, but the **reason is different**.

```text
loc  = LABEL
iloc = POSITION
```

# Module 8 — Boolean Filtering

Filtering is one of the most important Pandas skills.

```python
df[df["Age"] > 20]
```

For multiple conditions, use:
- `&` for AND
- `|` for OR
- `~` for NOT

Each condition should be enclosed in parentheses.

Do **not** use Python's `and` / `or` between Series conditions.

## 🔎 Detailed Explanation — Module 8

Boolean filtering builds a True/False mask and uses it to keep only rows where the condition is True. A condition such as `df['Age'] > 20` produces a Boolean Series aligned with the DataFrame index. Applying that mask returns the matching rows.

For multiple conditions, use `&` for AND and `|` for OR, and put each condition inside parentheses. Python's `and` and `or` are not the correct operators for element-wise Series comparisons. Filtering can also be combined with `isin`, `between`, string methods, and missing-value checks.

Good filtering is explicit about business logic. Write the condition so another reader can tell exactly which rows are intended to remain.

# Module 30 — Resampling and Rolling Windows

For time-indexed data:

- `resample()` groups observations into calendar/time bins.
- `rolling()` calculates moving-window statistics.

These are different concepts:

```text
resample → change the time frequency/grouping
rolling  → calculate over a moving window
```

## 🔎 Detailed Explanation — Module 30

Resampling groups time-indexed observations into calendar-based bins such as days, weeks, or months. Rolling windows instead calculate statistics over moving neighborhoods of observations. Both are useful for understanding time-dependent patterns.

Resampling changes the time frequency of the result and usually involves aggregation. Rolling calculations preserve a moving relationship between nearby observations, such as a moving average.

Window size, missing observations, and the alignment of the resulting index all matter. A rolling average is not the same as a monthly resample, even if both produce a smoother-looking series.

# Module 35 — Copy vs View and Safe Assignment

For teaching and professional code, the safest habit when creating an independently editable subset is:

subset = df.loc[condition].copy()

## 🔎 Detailed Explanation — Module 35

Copy-versus-view behavior is important because a DataFrame may be related to another object through shared underlying data or a derived selection. Ambiguous chained assignment can lead to confusing behavior and warnings.

For safe teaching and professional workflows, make ownership explicit. If you intend to create an independent working DataFrame, use `.copy()`. For assignment, prefer clear `.loc[...] = ...` patterns that state which rows and columns are being modified.

The goal is not to memorize every internal memory detail. The goal is to write code where it is obvious which object is being changed and to avoid accidental modification of an unintended object.

# 👀 Visual: Large-Data Thinking

```text
SMALL DATA
┌──────────────────────────────┐
│ load everything into memory  │
└──────────────────────────────┘

LARGE DATA
        ↓
   Can we load less?
        ↓
 columns / filtering / chunks
        ↓
   process efficiently
        ↓
   validate the result
```

Chunking:

```text
FILE
████████████████████████████
  ↓       ↓       ↓       ↓
chunk1  chunk2  chunk3  chunk4
  ↓       ↓       ↓       ↓
process process process process
  └───────┴───────┴───────┘
              ↓
           result
```

````python
# Reproducible chunking demo using a locally-created CSV.
large_demo = pd.DataFrame({
    "ID": range(1, 11),
    "Score": [55, 70, 80, 90, 65, 88, 92, 77, 61, 99]
})

large_path = "pandas_chunk_demo.csv"
large_demo.to_csv(large_path, index=False)

total = 0
for chunk in pd.read_csv(large_path, chunksize=3):
    total += chunk["Score"].sum()

assert total == large_demo["Score"].sum()
print("Chunked total:", total)
````

This is an advanced extension of the Pandas workflow. The goal is not to memorize another command; it is to understand **when the operation changes the shape, meaning, or presentation of your data**.

## 🧠 Mental model

Continuous values → pd.cut()
Fixed interval boundaries
Meaningful labels
Example: age groups

Continuous values → pd.qcut()
Quantile boundaries
Roughly balanced group sizes
Useful for ranks/segments

### Why this matters
These operations solve common real-data problems that appear after the basic DataFrame workflow is comfortable. Learn the question first, then choose the operation.

## 🖼️ Visual Explanation — Module 41

![Module 41 visual](pandas_visuals/module_41.png)

> **Look at the picture first.** Trace what happens to rows, columns, index levels, or values before reading the code.

## 🔧 Syntax and parameters

The exact parameter names matter. Read the call from left to right: **object → operation → important parameters → result**.

```python
pd.cut(x, bins=..., labels=..., include_lowest=...) / pd.qcut(x, q=..., labels=...)
```

### Worked example

The following example is intentionally small so you can predict the result before reading it.

````python
import pandas as pd

age = pd.Series([2, 8, 17, 23, 31, 47, 66], name="Age")

# cut: you define the boundaries
age_band = pd.cut(
    age,
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teen", "Adult", "Middle", "Senior"],
    include_lowest=True,
)
print(age_band)

# qcut: Pandas chooses quantile boundaries
quartile_band = pd.qcut(age, q=4, labels=["Q1", "Q2", "Q3", "Q4"])
print()
print(quartile_band)
````

## 🔍 What to notice

- **Shape:** Does the operation add rows, remove rows, move labels, or keep the same shape?
- **Meaning:** Does it change the data values, group them, or only change presentation?
- **Index:** Does the index become repeated or hierarchical?
- **Missing data:** Could unmatched, empty, or missing values appear?
- **Parameters:** Which parameter changes the default behavior?

### Common mistakes
- Using an operation without first deciding the analytical question.
- Confusing a display operation with a data transformation.
- Ignoring index levels after a reshape.
- Assuming a text pattern matches every row.
- Measuring memory without `deep=True` when object/string storage is important.

### When to use / when NOT to use
Use the operation when its mental model matches the problem. Do not use it simply because it is a newer or more advanced method. A simpler Pandas operation is often clearer.

# Module 54 — shift()

**Category:** Time Series & Lagged Features

Move values across rows while keeping the index in place.

# `shift()` — move values without moving the index

`shift()` moves a Series or DataFrame's **values** by a chosen number of periods. The labels stay where they are. This makes it a core building block for time-series comparisons and lagged features.

### Mental model

```text
Original:  A  B  C  D
shift(1):  NaN A  B  C
shift(-1): B  C  D  NaN
```

### Important parameters
- `periods=1`: one row down; negative values move upward.
- `freq=`: shifts **time labels** when working with a datetime index rather than simply moving values.
- `fill_value=`: optional value used for newly created gaps.

### Real-world use
Previous-day sales, previous month's balance, lagged sensor readings, and before/after comparisons.

# Module 80 — wide_to_long()

**Category:** Specialized Wide → Long Reshaping

Reshape repeated column-name patterns into a long format.

# `pd.wide_to_long()`

`wide_to_long()` is useful when column names encode a repeated variable pattern, such as `score_2025`, `score_2026`, `rank_2025`, `rank_2026`.

It is more specialized than `melt()`: use it when the column naming pattern itself carries the structure you want to extract.

Key parameters include `stubnames`, `i`, `j`, and `sep`/`suffix` depending on the schema.

## 🔎 Detailed Explanation — Module 80

**Think first:** What changes? What stays aligned? What question does this operation answer?

### Common mistakes to watch for
- Confusing a transformed value with a filtered row.
- Ignoring index alignment or ordering assumptions.
- Assuming missing values, ties, dtypes, or timezones behave like plain Python objects.
- Using a specialized operation when a simpler Pandas operation communicates the intent better.

### Professional habit
After running the operation, inspect the result's shape, labels, dtype, and a few representative values. Then explain the result in words before moving on.

````python
import pandas as pd

df = pd.DataFrame({
    "athlete": ["Ali", "Sara"],
    "score_2025": [80, 90],
    "score_2026": [85, 95]
})
long = pd.wide_to_long(df, stubnames="score", i="athlete", j="year", sep="_", suffix=r"\d+")
print(long.reset_index())
````
