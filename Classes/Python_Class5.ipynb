- **Git** = a tool that tracks changes in your project.
- **GitHub** = an online platform where Git repositories can be stored.

# Step 3: Run the Git Commands One by One

## 1. Start Git in the project

```bash
git init
```

### What does `git init` do?

It initializes Git in your current project folder.

In simple words:

> "Git, start tracking this project."

You normally run this once when starting Git for that project.

````python
# Start Git in this project
# The "!" runs this as a shell command instead of Python code.
!git init
````

### Line-by-line explanation

1. `# Start Git...` is a comment; Python (and the shell) ignore lines starting with `#`.
2. `!git init` — the `!` tells the notebook to send everything after it to the system shell. `git init` then creates a hidden `.git` folder in the current directory, which is how Git begins tracking this project.

## 2. Add all project files

```bash
git add .
```

### What does this mean?

`git add .` tells Git to prepare the files in the current project for the next commit.

Think:

> "Git, include my current project changes in the next saved version."

The `.` means the current folder and its contents.

````python
# Add the project files
!git add .
````

### Line-by-line explanation

1. `# Add the project files` is a comment.
2. `!git add .` stages every file in the current folder (`.`) so Git knows to include them in the next commit. Nothing is permanently saved yet — this step only prepares the files.

## 3. Create a commit

```bash
git commit -m "first commit"
```

### What is a commit?

A **commit** is a saved checkpoint of your project.

The part:

```bash
-m "first commit"
```

adds a message describing the commit.

Think:

> "Save a checkpoint of my project and call it `first commit`."

````python
# Save a checkpoint of the project
!git commit -m "first commit"
````

### Line-by-line explanation

1. `# Save a checkpoint...` is a comment.
2. `!git commit -m "first commit"` saves everything that was staged with `git add .` as a permanent checkpoint in the project's history. The `-m` flag lets you attach a short message (`"first commit"`) describing what this checkpoint contains.

# Step 4: Connect Your Local Project to GitHub

After creating the GitHub repository, GitHub provides a repository URL.

It will look similar to:

```text
https://github.com/USERNAME/RepoName.git
```

Then run:

```bash
git remote add origin https://github.com/USERNAME/RepoName.git
```

### What does this command do?

It connects your local Git project to the GitHub repository.

Think:

> "Git, this GitHub repository is the online home for this project."

### What is `origin`?

`origin` is simply the usual name Git gives to the remote repository.

````python
# Connect your local project to your GitHub repository.
# Replace the URL below with the URL of YOUR repository before running this.
!git remote add origin https://github.com/USERNAME/RepoName.git
````

### Line-by-line explanation

1. The first two lines are comments explaining what to do before running the command.
2. `!git remote add origin https://github.com/USERNAME/RepoName.git` tells Git that a remote (an online copy of the repository) named `origin` exists at the given GitHub URL. `USERNAME` and `RepoName` must be replaced with your actual GitHub username and repository name.

# Step 5: Upload the Project

First, set the main branch name:

```bash
git branch -M main
```

Then push your project:

```bash
git push -u origin main
```

### What does `git push` do?

`push` sends your committed changes from your computer to the remote GitHub repository.

Think:

> **commit = save a checkpoint locally**  
> **push = send that checkpoint to GitHub**

The `-u` connects the local `main` branch with the remote `main` branch, making future pushes easier.

````python
# Set the main branch
!git branch -M main

# Upload the project to GitHub
!git push -u origin main
````

### Line-by-line explanation

1. `!git branch -M main` renames (or creates) the current branch and names it `main`. This matches GitHub's default branch name.
2. `!git push -u origin main` uploads all committed checkpoints to the `origin` remote (your GitHub repository), on the `main` branch. The `-u` flag remembers this connection, so future runs can simply use `!git push`.

# Complete Command Sequence

After creating the GitHub repository and opening your project folder in the terminal:

```bash
git init
git add .
git commit -m "first commit"
git remote add origin https://github.com/USERNAME/RepoName.git
git branch -M main
git push -u origin main
```

Run them **one at a time**, in this exact order.



*The five-step Git workflow, start to finish.*
