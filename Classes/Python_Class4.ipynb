# 1. What is a List?

A **list** is a Python data structure used to store multiple values inside one variable.

Instead of creating separate variables:

```python
city_1 = "Atlanta"
city_2 = "Baltimore"
city_3 = "Chicago"
```

we can put the values together:

```python
cities = ["Atlanta", "Baltimore", "Chicago"]
```

A list:
- keeps items in an order,
- can contain different data types,
- can be changed after creation,
- can contain duplicate values,
- uses indexes starting from `0`.

````python
city_1 = "Atlanta"
city_2 = "Baltimore"
city_3 = "Chicago"

cities = ["Atlanta", "Baltimore", "Chicago"]

print(cities)
````

### Line-by-line explanation

**Line 1**
```python
city_1 = "Atlanta"
```
Creates a variable called `city_1` and stores the string `"Atlanta"`.

**Line 2**
```python
city_2 = "Baltimore"
```
Creates another variable and stores `"Baltimore"`.

**Line 3**
```python
city_3 = "Chicago"
```
Creates another variable and stores `"Chicago"`.

**Line 5**
```python
cities = ["Atlanta", "Baltimore", "Chicago"]
```
Creates one list called `cities`. The square brackets `[]` tell Python that we are creating a list.

**Line 7**
```python
print(cities)
```
Displays the complete list.

# 2. Indexing: Accessing Items

Python gives every list item a position called an **index**.

Important:

- First item → index `0`
- Second item → index `1`
- Third item → index `2`

For a list of 6 items, the indexes are:

| Item position | Positive index | Negative index |
|---|---:|---:|
| 1st | 0 | -6 |
| 2nd | 1 | -5 |
| 3rd | 2 | -4 |
| 4th | 3 | -3 |
| 5th | 4 | -2 |
| 6th | 5 | -1 |

Negative indexing starts from the end.



*Every list item has both a positive and a negative index.*

````python
cities = ["Atlanta", "Baltimore", "Chicago", "Denver", "Los Angeles", "Seattle"]

print(cities[0])
print(cities[2])
print(cities[-1])
````

### Line-by-line explanation

```python
cities[0]
```
Gets the first item because Python starts counting indexes from `0`.

```python
cities[2]
```
Gets the third item.

```python
cities[-1]
```
Gets the last item. `-1` always means the last item, `-2` means the second-last item, and so on.

# 3. Changing a List Item

Lists are **mutable**, which means we can change their items after creating the list.

The basic pattern is:

```python
list_name[index] = new_value
```

````python
cities = ["Atlanta", "Baltimore", "Chicago"]

cities[1] = "Boston"

print(cities)
````

### Line-by-line explanation

```python
cities[1] = "Boston"
```
The item at index `1` (`"Baltimore"`) is replaced with `"Boston"`.

The original list is changed directly.

# 4. Slicing a List

**Slicing** means taking a portion of a list.

The general syntax is:

```python
list[start:stop:step]
```

- `start` = where to begin
- `stop` = where to stop (**stop is excluded**)
- `step` = how many positions to move each time

Examples:
- `cities[:4]` → from the beginning up to, but not including, index `4`
- `cities[2:5]` → index `2` through index `4`
- `cities[-3:]` → last three items
- `cities[::2]` → every second item
- `cities[::-1]` → reverse the list

````python
cities = ["Atlanta", "Baltimore", "Chicago", "Denver", "Los Angeles", "Seattle"]

print(cities[:4])
print(cities[2:4])
print(cities[-3:-1])
print(cities[-3:])
print(cities[::2])
print(cities[::-1])
````

### Line-by-line explanation

```python
cities[:4]
```
No `start` is given, so Python starts at the beginning. It stops before index `4`.

```python
cities[2:4]
```
Starts at index `2` and stops before index `4`.

```python
cities[-3:-1]
```
Starts at the third item from the end and stops before the last item.

```python
cities[-3:]
```
Starts at the third item from the end and continues to the end.

```python
cities[::2]
```
No start or stop is specified. The step is `2`, so Python takes every second item.

```python
cities[::-1]
```
A step of `-1` moves backwards, so the list is returned in reverse order.

# 5. Finding the Length and Checking Membership

Two very useful operations:

### `len()`
Returns the number of items in a list.

### `in`
Checks whether a value exists in a list and returns `True` or `False`.

````python
cities = ["Atlanta", "Baltimore", "Chicago", "Denver"]

print(len(cities))
print("Chicago" in cities)
print("Miami" in cities)
print("Miami" not in cities)
````

### Line-by-line explanation

```python
len(cities)
```
Counts how many items are in the list.

```python
"Chicago" in cities
```
Checks whether `"Chicago"` exists. The result is `True`.

```python
"Miami" in cities
```
Checks whether `"Miami"` exists. The result is `False`.

```python
"Miami" not in cities
```
Checks whether `"Miami"` does not exist. The result is `True`.

# 6. Adding Items

### `append()`
Adds **one item** to the end of a list.

### `insert()`
Adds **one item** at a specific index.

### `extend()`
Adds multiple items from another iterable, commonly another list.

````python
cities = ["Atlanta", "Baltimore", "Chicago"]

cities.append("Seattle")
print(cities)

cities.insert(1, "Boston")
print(cities)

cities.extend(["Denver", "Miami"])
print(cities)
````

### Line-by-line explanation

```python
cities.append("Seattle")
```
Adds `"Seattle"` to the end.

```python
cities.insert(1, "Boston")
```
Inserts `"Boston"` at index `1`. Existing items are shifted to the right.

```python
cities.extend(["Denver", "Miami"])
```
Adds both `"Denver"` and `"Miami"` as separate list items.

### Important difference

```python
cities.append(["Denver", "Miami"])
```
would add the **whole list as one item**.

```python
cities.extend(["Denver", "Miami"])
```
adds `"Denver"` and `"Miami"` separately.

# 7. Adding Lists with `+`

Lists can also be combined using `+`.

This creates a **new list**; it does not modify the original list unless we assign the result back.

````python
cities = ["Atlanta", "Baltimore", "Chicago"]

new_cities = cities + ["Dallas"]

print(cities)
print(new_cities)

cities = cities + ["Dallas"]

print(cities)
````

### Line-by-line explanation

```python
new_cities = cities + ["Dallas"]
```
Creates a new list containing the old cities plus `"Dallas"`. The original `cities` remains unchanged.

```python
cities = cities + ["Dallas"]
```
Creates the combined list and then stores that new list back in `cities`.

# 8. Removing Items

There are three important ways to remove items.

### `remove(value)`
Removes the **first matching value**.

### `del`
Deletes an item using its index.

### `pop(index)`
Removes an item and **returns the removed value**.

If no index is supplied to `pop()`, it removes the last item.

````python
cities = ["Atlanta", "Boston", "Chicago", "Denver", "Chicago"]

cities.remove("Chicago")
print(cities)

del cities[0]
print(cities)

removed_city = cities.pop()
print(removed_city)
print(cities)
````

### Line-by-line explanation

```python
cities.remove("Chicago")
```
Finds the first `"Chicago"` and removes it.

```python
del cities[0]
```
Deletes the item at index `0`.

```python
removed_city = cities.pop()
```
Removes the last item and stores that removed item in `removed_city`.

```python
print(removed_city)
```
Shows which item was removed.

```python
print(cities)
```
Shows the list after removal.

# 9. `pop()` with an Index

You can also tell `pop()` exactly which item to remove.

````python
tasks = ["running", "breakfast", "studying", "swimming"]

removed_task = tasks.pop(1)

print("Removed:", removed_task)
print("Remaining:", tasks)
````

### Line-by-line explanation

```python
removed_task = tasks.pop(1)
```
Removes the item at index `1` and stores it in `removed_task`.

Because `pop()` returns the removed value, we can reuse that value.

This is especially useful when you want to **remove something and do something with it**.

# 10. Useful List Methods

Here are several methods worth knowing:

| Method | Purpose |
|---|---|
| `clear()` | Removes all items |
| `count(value)` | Counts occurrences of a value |
| `index(value)` | Finds the first index of a value |
| `sort()` | Sorts the list |
| `reverse()` | Reverses the list |

These methods usually modify the existing list.

````python
numbers = [4, 2, 4, 1, 3]

print(numbers.count(4))
print(numbers.index(1))

numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)

numbers.clear()
print(numbers)
````

### Line-by-line explanation

```python
numbers.count(4)
```
Counts how many times `4` appears.

```python
numbers.index(1)
```
Returns the first index where `1` appears.

```python
numbers.sort()
```
Sorts the list in ascending order.

```python
numbers.reverse()
```
Reverses the current order.

```python
numbers.clear()
```
Removes every item, leaving an empty list `[]`.

# 11. A Practical Example: Moving Tasks

We can combine list methods to model a simple task workflow.

We have:
- `tasks` → tasks that are still pending
- `tasks_done` → tasks that have been completed

````python
tasks = ["praying Fajr", "running", "breakfast", "studying"]
tasks_done = []

completed_task = tasks.pop(0)
tasks_done.append(completed_task)

print("Pending:", tasks)
print("Done:", tasks_done)
````

### Line-by-line explanation

```python
tasks = [...]
```
Creates a list of pending tasks.

```python
tasks_done = []
```
Creates an empty list where completed tasks will be stored.

```python
completed_task = tasks.pop(0)
```
Removes the first task and saves it in `completed_task`.

```python
tasks_done.append(completed_task)
```
Adds that completed task to the `tasks_done` list.

The important idea is:

**remove from one list → store the value → add it to another list**

# 12. Assignment vs Shallow Copy vs Deep Copy

This is an important concept because beginners often think that every way of copying a list creates a completely independent list.

It does not.

There are three different ideas:

### A. Assignment
```python
b = a
```
Both variables refer to the **same list**.

### B. Shallow copy
```python
b = a.copy()
```
or
```python
b = a[:]
```
creates a new outer list, but nested mutable objects inside it can still be shared.

### C. Deep copy
```python
import copy
b = copy.deepcopy(a)
```
creates a fully independent copy, including nested objects.

# 13. Assignment: No New List

````python
a = [1, 2, 3]
b = a

b[0] = 99

print(a)
print(b)
````

### Line-by-line explanation

```python
a = [1, 2, 3]
```
Creates one list.

```python
b = a
```
Does **not** create another independent list. `b` refers to the same list as `a`.

```python
b[0] = 99
```
Changes the shared list.

Therefore both `a` and `b` show `[99, 2, 3]`.

# 14. Shallow Copy

A shallow copy creates a **new outer list**.

For a simple list containing immutable values such as integers, this often behaves exactly as beginners expect.

However, nested lists or other mutable objects can still be shared.

````python
a = [1, 2, 3]
b = a.copy()

b[0] = 99

print(a)
print(b)
````

### Line-by-line explanation

```python
a = [1, 2, 3]
```
Creates the original list.

```python
b = a.copy()
```
Creates a new outer list.

```python
b[0] = 99
```
Changes only `b` because the top-level integer item is not shared as a mutable object.

So:
- `a` stays `[1, 2, 3]`
- `b` becomes `[99, 2, 3]`

### Important correction

`list.copy()` is **not a deep copy**. It is a **shallow copy**.

# 15. Why Shallow Copy Matters with Nested Lists

Now look at a list containing another list.



*Shallow copy shares nested objects; deep copy makes everything independent.*

````python
a = [[1, 2], 3]
b = a.copy()

b[0][0] = 99

print(a)
print(b)
````

### Line-by-line explanation

```python
a = [[1, 2], 3]
```
The first item is itself a list: `[1, 2]`.

```python
b = a.copy()
```
Creates a new outer list, but the inner list `[1, 2]` is still shared.

```python
b[0][0] = 99
```
Changes the first item inside the shared inner list.

Therefore both `a` and `b` show the changed nested value.

# 16. Deep Copy

A deep copy creates a new outer list **and recursively copies nested objects**.

Use the `copy` module:

````python
import copy

a = [[1, 2], 3]
b = copy.deepcopy(a)

b[0][0] = 99

print(a)
print(b)
````

### Line-by-line explanation

```python
import copy
```
Imports Python's built-in `copy` module.

```python
a = [[1, 2], 3]
```
Creates a nested list.

```python
b = copy.deepcopy(a)
```
Creates a fully independent copy, including the nested list.

```python
b[0][0] = 99
```
Changes the nested list inside `b`.

Because the nested list is independent, `a` remains unchanged.
