# 1. 🏁 What is Python?

**Python** is a high-level, interpreted, general-purpose programming language.

It was created by **Guido van Rossum** and first released in **1991**.

Python is popular because its syntax is relatively easy to read and write.

### In simple words

Python is a programming language that lets us give instructions to a computer.

### Where can Python be used?

Python can be used for many different types of projects, including:

- Web development
- Data analysis
- Artificial intelligence
- Automation
- Machine learning
- Cloud and other technical applications

### Example idea

> **Python is a beginner-friendly programming language that can be used to build many different kinds of software.**



*Python is a single language used across many different fields.*

## 🧠 Understanding the word "programming language"

A **programming language** is a way for humans to write instructions that a computer can execute.

For example:

```python
print("Hello")
```

This tells Python to display `Hello` on the screen.

### Key words to remember

- **Python** → the programming language
- **Code** → instructions written for the computer
- **Program** → a set of instructions that performs a task

# 3. 🧩 Features of Python

Here are some important features of Python:

| Feature | Simple explanation |
|---|---|
| **Easy to Learn** | Python has a relatively simple and readable syntax. |
| **Interpreted** | Python code is executed by the Python interpreter. |
| **Object-Oriented** | Python supports programming with classes and objects. |
| **Portable** | Python can run on different operating systems such as Windows, macOS, and Linux. |
| **Rich Libraries** | Python has many built-in modules and a large ecosystem of libraries. |

### ⭐ Beginner focus

For now, remember these five ideas:

**Easy to learn → Interpreted → Object-oriented → Portable → Rich libraries**

## 🔹 Setting Up VS Code for Python

After installing both Python and VS Code:

### Step 1 — Open VS Code

Launch Visual Studio Code.

### Step 2 — Install Python extensions

Open the **Extensions** section from the left sidebar and search for:

- **Python**
- **Python Debugger**
- **Python Environment Manager**

Install the extensions needed for your Python setup.

### Step 3 — Create a project folder

Create a folder on your computer, for example on your **C:** or **D:** drive.

### Step 4 — Open the folder in VS Code

In VS Code:

**File → Open Folder → Select your project folder**

### Step 5 — Create your first Python file

Inside the folder, create either:

```text
first_learning_class.py
```

or

```text
first_learning_class.ipynb
```

### What do `.py` and `.ipynb` mean?

- **`.py`** → a Python script file
- **`.ipynb`** → a Jupyter Notebook file

Both can be used to work with Python, but they provide different working environments.

# 5. 🖥️ Writing Your First Python Code

Now let's write our first Python program.

The `print()` function displays information in the output.

For example:

```python
print("Your name")
```

Let's try it with a real name:

````python
print("My name is Saad.")
````

### 🔍 Line-by-line explanation

```python
print("My name is Saad.")
```

- `print` → tells Python to display something.
- `()` → parentheses are used to provide information to the `print()` function.
- `"My name is Saad."` → this is text, also called a **string**.
- The quotation marks tell Python that the content is text.

### Expected output

```text
My name is Saad.
```
