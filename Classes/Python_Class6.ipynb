## 2. Tuples

A **tuple** is similar to a list because it can store multiple values.

The main difference in this class is the syntax:

- List → `[]`
- Tuple → `()`

Example:

```python
tp_bio_data = ('Saad', 'is', 'a', 'Trainer')
```

Tuples are commonly used when the collection should not be changed accidentally, because tuples are **immutable** (their items cannot be modified after creation).

### Important: one-item tuple

This is **not** a tuple:

```python
tp = (1)
```

Python treats `(1)` as just the number `1`.

For a one-item tuple, add a comma:

```python
tp = (1,)
```

````python
tp_bio_data = ('Saad', 'is', 'a', 'Trainer', 'of', 'python_programming')

print(type(tp_bio_data))
print(tp_bio_data[1])
````

### Line-by-line explanation

1. `tp_bio_data = (...)` creates a tuple with six items, using round brackets instead of square brackets.
2. `print(type(tp_bio_data))` displays `<class 'tuple'>`, confirming it is a tuple.
3. `print(tp_bio_data[1])` accesses the item at index `1` (the second item), which is `'is'`. Tuples support indexing just like lists.

````python
lst = [1]
print(type(lst))

tp = (1,)
print(type(tp))
````

### Line-by-line explanation

1. `lst = [1]` creates a list with one item; `type(lst)` correctly shows `<class 'list'>`.
2. `tp = (1,)` creates a tuple with one item **because of the trailing comma**; `type(tp)` shows `<class 'tuple'>`.
3. If the comma were removed (`tp = (1)`), Python would treat `(1)` as just the number `1` in parentheses, and `type(tp)` would show `<class 'int'>` instead of `<class 'tuple'>`.

### Tuple methods

A tuple supports methods such as:

- `.count()` → counts how many times a value appears
- `.index()` → finds the index of the first matching value

````python
print(tp_bio_data.count('Saad'))
print(tp_bio_data.index('Saad'))

personal_bio = ('Saad', 18, '5.9 feet', 17)
print(personal_bio.index(17))
print(personal_bio.count('Saad'))
````

### Line-by-line explanation

1. `tp_bio_data.count('Saad')` counts how many times `'Saad'` appears in `tp_bio_data`. It appears once, so this prints `1`.
2. `tp_bio_data.index('Saad')` finds the position of the first `'Saad'`, which is index `0`.
3. `persona_bio = (...)` creates a new tuple mixing text and numbers; tuples can hold different data types together, just like lists.
4. `persona_bio.index(17)` finds the **first** index where `17` appears. Even though `17` appears twice (index `1` and index `3`), `.index()` only returns the first match, which is `1`.
5. `persona_bio.count('Saad')` counts how many times `'Saad'` appears, which is `1`.

### Tuple unpacking

You can put tuple values into separate variables:

```python
name, city, age = person
```

Python assigns the values from left to right.

So:

- `name` gets the first value
- `city` gets the second value
- `age` gets the third value

The number of variables must match the number of values.

````python
person = ("Saad", "Karachi", 18)

name, city, age = person

print(name)
print(city)
print(age)
````

### Line-by-line explanation

1. `person = ("Saad", "Karachi", 17)` creates a tuple with three values.
2. `name, city, age = person` unpacks the tuple: `name` receives `"Saad"` (position 0), `city` receives `"Karachi"` (position 1), and `age` receives `17` (position 2).
3. The three `print()` calls display each unpacked variable. If you tried to unpack into more or fewer variables than there are values, Python would raise a `ValueError`.

## 3. `for` Loops

A `for` loop repeats code for every item in a collection.

Basic structure:

```python
for temporary_variable in collection:
    code_to_repeat
```

Think of it as:

> "Take one item at a time from this collection and run the code."

The variable name can be anything, but a meaningful name makes the code easier to understand.



*How a for loop repeats until every item has been visited.*

````python
fruits = ['Apple', 'Mango', 'Grapes', 'Banana']

for fruit in fruits:
    print(fruit)
````

### Line-by-line explanation

```python
for fruit in fruits:
```
Take one fruit from `fruits` at a time and temporarily store it in `fruit`.

```python
print(fruit)
```
Print the current fruit.

The loop continues until every fruit has been processed.

````python
for data in bio_data:
    print(data)
````

### Line-by-line explanation

1. `for data in bio_data:` visits each item in the `bio_data` list created earlier in this notebook, one at a time, storing it in `data`.
2. `print(data)` displays the current item.
3. The loop repeats automatically for every item in the list; there is no need to know the list's length in advance.

## 4. `enumerate()`

Normally, a `for` loop gives you the item.

`enumerate()` gives you **both the index and the item**.

Example:

```python
for index, value in enumerate(fruits):
    print(index, value)
```

The result conceptually looks like:

```text
0 Apple
1 Mango
2 Grapes
3 Banana
```

````python
fruits = ['Apple', 'Mango', 'Grapes', 'Banana']

for index, fruit in enumerate(fruits):
    print(index, fruit)
````

````python
for data in enumerate(fruits):
    print(f"{data[1]} is at index {data[0]}")
````

### Line-by-line explanation

1. `enumerate(fruits)` produces pairs of `(index, value)` for every item in `fruits`.
2. In the first example, `for index, fruit in enumerate(fruits):` directly unpacks each pair into two variables — `index` and `fruit` — so `print(index, fruit)` shows the position and the value together.
3. In the second example, `for data in enumerate(fruits):` does **not** unpack the pair; instead, `data` holds the whole pair, such as `(0, 'Apple')`.
4. `data[0]` then accesses the index, and `data[1]` accesses the value, because each pair behaves like a small tuple.
5. The f-string `f"{data[1]} is at index {data[0]}"` inserts the value and index into a sentence.

### Why `data[0]` and `data[1]`?

Each result from `enumerate()` is a pair:

```python
(index, value)
```

So:

- `data[0]` → index
- `data[1]` → value

For beginners, the first version (`for index, fruit in enumerate(fruits):`) is usually easier to read.

## 5. `zip()`

`zip()` combines items from two or more collections **position by position**.

Example:

```python
zip([1, 2, 3], ['A', 'B', 'C'])
```

pairs them as:

```text
(1, 'A')
(2, 'B')
(3, 'C')
```

### Important
If the collections have different lengths, `zip()` stops when the **shortest collection** runs out of items.

````python
fruits_index = [2, 1, 4, 3, 5]
fruits = ['Apple', 'Mango', 'Grapes', 'Banana']

for index, fruit in zip(fruits_index, fruits):
    print(index, fruit)
````

### Line-by-line explanation

1. `fruits_index` contains five numbers, while `fruits` contains four fruits.
2. `zip(fruits_index, fruits)` pairs up items position by position: the 1st number with the 1st fruit, the 2nd number with the 2nd fruit, and so on.
3. `zip()` stops as soon as the **shorter** collection (`fruits`, with 4 items) runs out, so only four pairs are produced. The extra `5` in `fruits_index` has no matching fruit and is simply ignored.
4. `for index, fruit in zip(...):` unpacks each pair into `index` and `fruit`, and `print(index, fruit)` displays them together.

This is why `zip()` should not be described as "always using every item" — it only uses as many items as the shortest collection provides.

## 6. `range()`

`range()` generates a sequence of numbers, commonly for loops.

### `range(stop)`

```python
range(5)
```

produces:

```text
0, 1, 2, 3, 4
```

The stop value is **not included**.

### `range(start, stop)`

```python
range(2, 5)
```

produces:

```text
2, 3, 4
```

### `range(start, stop, step)`

```python
range(2, 5, 2)
```

produces:

```text
2, 4
```

````python
for number in range(5):
    print(number)

print("---")

for number in range(2, 5):
    print(number)

print("---")

for number in range(2, 5, 2):
    print(number)
````

### Line-by-line explanation

1. `for number in range(5):` produces the numbers `0, 1, 2, 3, 4` (five numbers, starting at `0`, stopping **before** `5`). Each is printed on its own line.
2. `print("---")` prints a separator line so the three groups of output are easy to tell apart.
3. `for number in range(2, 5):` produces `2, 3, 4` — it starts at `2` and stops before `5`.
4. `for number in range(2, 5, 2):` produces `2, 4` — it starts at `2`, stops before `5`, and moves in steps of `2` (skipping `3`).

### Repeating something a fixed number of times

If you want to print `"apple"` five times:

```python
for _ in range(5):
    print("apple")
```

The `_` means we do not need the loop number.

````python
for _ in range(5):
    print('apple')
````

### Line-by-line explanation

1. `for _ in range(5):` repeats the loop body five times. The underscore `_` is a variable name too, but by convention it signals "this loop variable is not used inside the loop."
2. `print('apple')` runs once for each of the five repetitions, so `apple` is printed five times.

````python
print("Apples\n" * 5)
````

### Line-by-line explanation

1. `"Apples\n"` is a String containing the word `Apples` followed by a newline character (`\n`), which moves the cursor to the next line.
2. Multiplying a String by a number (`"Apples\n" * 5`) repeats that String five times in a row, producing `"Apples\nApples\nApples\nApples\nApples\n"`.
3. `print(...)` displays the result, and because each repetition ends with `\n`, the word `Apples` appears on five separate lines — a loop is not even required for this particular pattern.

## 7. Nested `for` Loops

A **nested loop** is a loop inside another loop.

The outer loop runs first. For each outer-loop value, the inner loop runs through all of its values.

This is useful when you need to compare every item from one collection with every item from another collection.

````python
group_A = ['Karachi', 'Lahore', 'Islamabad']
group_B = []

user_input = input("Enter your home city for the cricket match: ")

if user_input in group_A:
    print("Your city is already in group_A.")
else:
    group_B.append(user_input)

print("Group A:", group_A)
print("Group B:", group_B)
````

### Line-by-line explanation

1. `group_A = [...]` creates a list of three cities. `group_B = []` creates an empty list.
2. `user_input = input(...)` asks the user to type a city name and stores it as text.
3. `if user_input in group_A:` checks whether the typed city is already inside `group_A`.
4. If it is, the program prints a message and does nothing else.
5. If it is **not** (the `else` branch), `group_B.append(user_input)` adds the new city to `group_B`.
6. The final two `print()` calls show both groups so you can see the result.
7. **Note:** the nested-loop example below assumes `group_B` already has at least one item in it — run this cell first and enter a city that is *not* in `group_A` so the next example has something to compare.

````python
for a in range(len(group_A)):
    for b in range(len(group_B)):
        print(group_A[a], 'VS', group_B[b])

        if group_A[a] == group_B[b]:
            print("Matched")
            break
````

### Line-by-line explanation

1. `range(len(group_A))` produces the valid **indexes** of `group_A`: `0`, `1`, `2` (since it has 3 items).
2. The outer loop `for a in range(len(group_A)):` visits each of those indexes one at a time.
3. For every single value of `a`, the inner loop `for b in range(len(group_B)):` runs completely through all indexes of `group_B`.
4. `print(group_A[a], 'VS', group_B[b])` prints every possible pairing between a city from `group_A` and a city from `group_B`.
5. `if group_A[a] == group_B[b]:` checks whether the two cities are exactly the same text.
6. If they match, `"Matched"` is printed and `break` stops the **inner** loop only — the outer loop continues to its next value of `a`.
7. If `group_B` is empty (no items were added in the previous cell), the inner loop never runs, so nothing is printed here — this is expected, not an error.

## 7.1 The `continue` Statement

`break` completely stops a loop.

`continue` is different: it only skips the **current** repetition and moves on to the next one. The loop itself keeps running.

```python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

This prints `1, 2, 4, 5` — the number `3` is skipped, but the loop continues afterward.

````python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
````

### Line-by-line explanation

1. `for number in range(1, 6):` loops through `1, 2, 3, 4, 5`.
2. `if number == 3:` checks whether the current number is `3`.
3. `continue` — when this line runs, Python immediately jumps back to the top of the loop and moves on to the **next** value, skipping everything below it for this repetition. So `print(number)` never runs when `number` is `3`.
4. For every other number, the `if` condition is `False`, so `continue` does not run, and `print(number)` displays the number normally.
5. The output is `1`, `2`, `4`, `5` — notice `3` is missing, but the loop does not stop; it keeps going all the way to `5`.

### `break` vs `continue`

| Statement | Effect |
|---|---|
| `break` | stops the loop completely |
| `continue` | skips only the current repetition, loop keeps going |

## 8. `while` Loops

A `while` loop repeats code **as long as its condition is `True`**.

Basic structure:

```python
while condition:
    code_to_repeat
```

When the condition becomes `False`, the loop stops.

### Very important
Make sure something inside the loop can eventually make the condition false. Otherwise, you may create an infinite loop.

````python
number = 1
total = 0

while number <= 5:
    total = total + number
    print("Adding:", number)
    number = number + 1

print("Final Sum:", total)
````

### Line-by-line explanation

```python
number = 1
```
Start counting from 1.

```python
total = 0
```
Start the total at 0.

```python
while number <= 5:
```
Keep repeating while `number` is 5 or less.

```python
total = total + number
```
Add the current number to the total.

```python
number = number + 1
```
Increase the number by 1 so the loop can eventually stop.

The final answer is:

```text
Final Sum: 15
```

````python
num = 2
i = 1

while i <= 10:
    result = num * i
    print(num, "x", i, "=", result)
    i = i + 1

print("Table completed")
````

### Line-by-line explanation

1. `num = 2` is the number whose multiplication table we want; `i = 1` is the counter, starting at `1`.
2. `while i <= 10:` keeps the loop running as long as `i` is `10` or less, so the table has exactly 10 rows.
3. `result = num * i` calculates the current row of the table (`2 x 1`, `2 x 2`, and so on).
4. `print(num, "x", i, "=", result)` displays that row in a readable sentence.
5. `i = i + 1` increases the counter so the loop eventually reaches `11` and stops.
6. `print("Table completed")` runs once, after the loop has fully finished.

## 9. Country checking with a `for` loop

This example checks whether the user's input matches a country in a list.

The important ideas here are:

- `input()` gets text from the user.
- `for` checks countries one by one.
- `==` compares two values.
- `break` stops the loop after a match.

````python
clean_countries = ['Singapore', 'Japan', 'THE GREAT AMERICA', 'China']

user_input = input("Enter a country name that you want to check: ")

for country in clean_countries:
    print(f"Checking: {country}")

    if user_input == country:
        print(f"{user_input} is in the clean countries list.")
        break
````

### Line-by-line explanation

1. `clean_countries = [...]` creates a list of four country names.
2. `user_input = input(...)` asks the user to type a country and stores it as text.
3. `for country in clean_countries:` visits each country in the list, one at a time.
4. `print(f"Checking: {country}")` shows which country is currently being compared, which helps you see the loop working step by step.
5. `if user_input == country:` compares the user's text exactly with the current country (comparisons are case-sensitive, so `"japan"` would not match `"Japan"`).
6. If they match, a confirmation message is printed and `break` immediately stops the loop — there is no need to keep checking the remaining countries.
7. If none of the countries match, the loop simply finishes after checking all of them, and no match message is printed.

### Important note about comparing two identical strings

The expression:

```python
'THE GREAT AMERICA' == 'THE GREAT AMERICA'
```

is simply a comparison. It returns `True`.

It is **not connected to the loop above** unless you actually use its result inside the loop's logic — writing this comparison on its own line, by itself, does nothing more than produce a Boolean value.

````python
print('THE GREAT AMERICA' == 'THE GREAT AMERICA')
````

### Line-by-line explanation

1. `'THE GREAT AMERICA' == 'THE GREAT AMERICA'` compares two identical Strings, character by character, and evaluates to `True` because every character matches.
2. `print(...)` displays that Boolean result, `True`, on its own — this line is a standalone demonstration, separate from the country-checking loop above.

## 10. Country checking with a `while` loop

A `while` loop is useful when we want to keep asking the user until a condition is satisfied.

In this version, the loop stops when the user enters a country from the list.

````python
clean_countries = ['Singapore', 'Japan', 'THE GREAT AMERICA', 'China']

while True:
    user_input = input("Enter a country to check: ")

    if user_input in clean_countries:
        print(f"{user_input} is in the clean countries list.")
        break
````

### Line-by-line explanation

1. `clean_countries = [...]` recreates the list of valid countries.
2. `while True:` starts a loop whose condition is always `True`, meaning it would normally repeat forever unless something inside stops it.
3. `user_input = input(...)` asks the user for a country every time the loop runs.
4. `if user_input in clean_countries:` checks whether the typed country exists anywhere in the list.
5. If it does, a message is printed and `break` exits the loop — this is what eventually stops the "infinite" `while True:` loop.
6. If the input does not match, the loop goes back to the top and asks again, repeating indefinitely until a valid country is entered.

### Using `q` to stop

Another common pattern keeps asking until the user enters exactly `q`.

Notice that:

```python
usr_input != 'q'
```

means "the input is not equal to `q`."

So any input other than lowercase `q` keeps the loop running.

`Q`, ` q`, and `q ` are different strings from `q`, because String comparison is case-sensitive and sensitive to extra spaces.

````python
clean_countries = ['Singapore', 'Japan', 'THE GREAT AMERICA', 'China']

user_input = ''

while user_input != 'q':
    user_input = input("Enter a country to check, or q to quit: ")

    if user_input in clean_countries:
        print(f"{user_input} is in the clean countries list.")
````

### Line-by-line explanation

1. `user_input = ''` creates the variable with an empty String **before** the loop starts. This is necessary so that `while user_input != 'q':` has something to check the very first time — an empty String is not equal to `'q'`, so the loop begins.
2. `while user_input != 'q':` keeps repeating as long as the user has not typed exactly `q`.
3. `user_input = input(...)` asks for input and **overwrites** `user_input` with whatever the user just typed.
4. `if user_input in clean_countries:` checks the new input against the list and prints a message if it matches.
5. The loop then goes back to its condition; if the user typed `q`, the loop stops. Otherwise it repeats.

## 11. String Casing

The following methods can be used on strings:

- `.upper()` → converts letters to uppercase
- `.lower()` → converts letters to lowercase
- `.title()` → capitalizes the first letter of each word

These methods return a **new** string; they do not change the original string automatically.

````python
text = "Saad"

print(text.upper())
print(text.lower())
print(text.title())
````

### Line-by-line explanation

1. `text = "Saad"` creates the original String.
2. `text.upper()` returns a new String with every letter converted to uppercase: `"SAAD"`.
3. `text.lower()` returns a new String with every letter converted to lowercase: `"saad"`.
4. `text.title()` returns a new String with the first letter of each word capitalized: `"Saad"` (already title case here, since it is a single word starting with a capital letter).
5. In every case, `text` itself is **unchanged** after these calls — each method returns a brand-new String rather than modifying the original.

## 11.1 More Useful String Methods

Beyond casing, a few more string methods come up constantly in real programs:

- `.strip()` → removes extra spaces (or other characters) from the start and end of a string
- `.split()` → breaks a string into a list of pieces, using a separator
- `.replace(old, new)` → returns a new string with every occurrence of `old` swapped for `new`
- `.find(value)` → returns the index where `value` first appears, or `-1` if it is not found

````python
sentence = "  Karachi, Lahore, Islamabad  "

print(sentence.strip())

cities = sentence.strip().split(", ")
print(cities)

print(sentence.replace("Lahore", "Multan"))

print(sentence.find("Lahore"))
print(sentence.find("Peshawar"))
````

### Line-by-line explanation

1. `sentence = "  Karachi, Lahore, Islamabad  "` creates a String with extra spaces at the start and end.
2. `sentence.strip()` returns a new String with those leading and trailing spaces removed, giving `"Karachi, Lahore, Islamabad"`. Note that it does **not** remove spaces in the middle of the text.
3. `sentence.strip().split(", ")` first strips the extra spaces, then `.split(", ")` breaks the remaining text into a list wherever it finds the separator `", "`, giving `['Karachi', 'Lahore', 'Islamabad']`.
4. `sentence.replace("Lahore", "Multan")` returns a new String with every occurrence of `"Lahore"` swapped for `"Multan"`. The original `sentence` variable itself is unchanged, exactly like `.upper()` and `.lower()` earlier.
5. `sentence.find("Lahore")` searches for `"Lahore"` inside `sentence` and returns the index where it starts.
6. `sentence.find("Peshawar")` returns `-1` because `"Peshawar"` does not appear anywhere in `sentence`.

## 12. Dictionaries

A **dictionary** stores data as **key-value pairs**.

Example:

```python
customer = {
    "first name": "Saad",
    "last name": "Ahmed"
}
```

Think of it like:

```text
key          value
first name → Saad
last name  → Ahmed
```

You access a value using its key:

```python
customer["first name"]
```

### Beginner note
Do not name your variable `dict`, because `dict` is already Python's built-in dictionary type. Use names such as `student`, `customer`, or `person`.



*A dictionary maps each key to exactly one value.*

````python
student = {
    'age': 15,
    'weight': 15
}

print(type(student))
print(student['age'])
````

### Line-by-line explanation

1. `student = {...}` creates a dictionary with two keys, `'age'` and `'weight'`, each holding the value `15`. Curly braces `{}` tell Python to create a dictionary.
2. `print(type(student))` displays `<class 'dict'>`.
3. `print(student['age'])` looks up the value stored under the key `'age'` and displays `15`. Square brackets after a dictionary name are used to access a value by its **key**, not by a numeric position like a list.

````python
customer = {
    "first name": "Saad",
    "last name": "Ahmed",
    "address": "North Karachi"
}

print(customer['address'])
````

### Line-by-line explanation

1. `customer = {...}` creates a dictionary with three keys: `"first name"`, `"last name"`, and `"address"`. Each key is separated from its value by a colon `:`, and each key-value pair is separated by a comma.
2. `print(customer['address'])` looks up the key `"address"` and displays its value, `"North Karachi"`.

> **Common beginner mistake to avoid:** every key-value pair inside a dictionary must end with a comma if another pair follows it. Forgetting a comma (for example writing two Strings next to each other with nothing between them) is a frequent source of confusing bugs, because Python may silently join adjacent String literals together instead of raising an error.

### Dictionary key/value flexibility

In Python:

- a key can be a number
- a value can be a number
- keys and values can have different types in the same dictionary

The key is what you use to access the value.

````python
dict_versatility1 = {
    1: "Muhammad",
    2: "Saad",
    3: "Ahmed"
}

print(dict_versatility1[2])
````

````python
dict_versatility2 = {
    "Muhammad": 1,
    "Saad": 2,
    "Ahmed": 3
}

print(dict_versatility2['Saad'])
````

````python
things_to_remember = {
    0: "the lowest number",
    "a dozen": 12,
    "snake eyes": "a pair of ones",
    13: "a baker's dozen"
}

things_to_remember
````

### Line-by-line explanation

1. `dict_versatility1` uses **numbers as keys** and names as values; `dict_versatility1[2]` looks up the key `2` and returns `"Saad"`.
2. `dict_versatility2` flips this around, using **names as keys** and numbers as values; `dict_versatility2['Saad']` looks up the key `'Saad'` and returns `2`.
3. `things_to_remember` mixes everything: some keys are numbers (`0`, `13`) and some are Strings (`"a dozen"`, `"snake eyes"`); some values are Strings and one value is a number (`12`). This shows that a single dictionary can freely mix data types for both its keys and its values.
4. Writing `things_to_remember` alone as the last line of the cell displays the whole dictionary automatically, the same way `bio_data` did earlier in this notebook.

## 13. Adding and changing dictionary values

To add a new key-value pair:

```python
customer["phone"] = "..."
```

If the key does not exist, Python adds it.

If the key already exists, Python **changes/updates** its value.

````python
customer = {
    "first name": "Saad",
    "last name": "Ahmed",
    "address": "North Karachi"
}

customer["phone"] = "+92023456778"
customer["city"] = "Karachi"

print(customer)
````

### Line-by-line explanation

1. `customer = {...}` creates a dictionary with three existing keys.
2. `customer["phone"] = "+92023456778"` — since `"phone"` does **not** already exist as a key, Python adds it as a brand-new key-value pair.
3. `customer["city"] = "Karachi"` similarly adds a new `"city"` key.
4. `print(customer)` shows the dictionary now containing all five key-value pairs.

````python
customer["city"] = "New York"
print(customer)
````

### Line-by-line explanation

1. `customer["city"] = "New York"` — because the key `"city"` **already exists** (it was just added in the previous cell), this line updates its value instead of creating a duplicate key.
2. `print(customer)` confirms that `"city"` now shows `"New York"` instead of `"Karachi"`; a dictionary key can only ever hold one current value at a time.

### Practical example: building a task dictionary

We can use a loop to build up a dictionary dynamically, based on however many tasks the user wants to enter.

````python
tasks = {}

task_count = int(input("How many tasks do you want to do today? "))

for count in range(task_count):
    task = input("Enter your task: ")
    tasks["Task " + str(count + 1)] = task

print(tasks)
````

### Line-by-line explanation

```python
tasks = {}
```
Creates an empty dictionary.

```python
task_count = int(input(...))
```
Gets the number of tasks from the user and converts it from text to an integer.

```python
for count in range(task_count):
```
Repeats once for each task.

```python
task = input(...)
```
Gets the task from the user.

```python
tasks["Task " + str(count + 1)] = task
```
Creates a key such as `Task 1`, `Task 2`, etc., and stores the user's task as its value. `str(count + 1)` converts the loop counter (which starts at `0`) into text so it can be joined with `"Task "` using `+`, and `+ 1` makes the numbering start from `1` instead of `0`.

## 15. Looping through dictionaries

There are three important ways shown in this class:

### Keys

```python
customer.keys()
```

### Values

```python
customer.values()
```

### Keys and values together

```python
customer.items()
```

````python
customer = {
    "first name": "Saad",
    "last name": "Ahmed",
    "address": "North Karachi",
    "city": "Karachi"
}

for key in customer.keys():
    print(key)
````

````python
for value in customer.values():
    print(value)
````

````python
for key, value in customer.items():
    print(key, "-->", value)
````

### Line-by-line explanation

1. `customer = {...}` recreates a dictionary with four key-value pairs.
2. `for key in customer.keys():` loops through only the **keys** (`"first name"`, `"last name"`, `"address"`, `"city"`), printing each one on its own line.
3. `for value in customer.values():` loops through only the **values** (`"Saad"`, `"Ahmed"`, `"North Karachi"`, `"Karachi"`), printing each one.
4. `for key, value in customer.items():` loops through **both** at once — each `.items()` pair is unpacked directly into `key` and `value` — so `print(key, "-->", value)` shows each key next to its matching value.

### Easy way to remember

| Method | Gives you |
|---|---|
| `.keys()` | keys |
| `.values()` | values |
| `.items()` | keys + values |

For most tasks, `.items()` is the most useful, since it gives you both pieces of information in one loop.

## 16. List of Dictionaries

A list can contain dictionaries.

This is very common when working with real-world data.

For example:

```python
[
    {"name": "Saad", "city": "Karachi"},
    {"name": "Abdul", "city": "Karachi"}
]
```

Think of it as:

- **list** → many records
- **dictionary** → information about one record

````python
list_of_dict = [
    {
        'first name': 'Saad',
        'last name': 'Ahmed',
        'address': 'North Karachi',
        'city': 'Karachi'
    },
    {
        'first name': 'Abdul',
        'last name': 'Qadir',
        'address': 'Power house',
        'city': 'Karachi'
    },
    {
        'first name': 'Safwan',
        'last name': 'Atif',
        'address': 'North Karachi',
        'city': 'Karachi'
    }
]

list_of_dict
````

### Line-by-line explanation

1. `list_of_dict = [...]` creates a **list** containing three **dictionaries**, each representing one person's information.
2. Each dictionary uses the same set of keys (`'first name'`, `'last name'`, `'address'`, `'city'`), which is a common and convenient pattern for storing multiple similar records.
3. Writing `list_of_dict` on the last line displays the entire structure — a list of three dictionaries — automatically.

````python
customer1 = list_of_dict[0]

print(customer1)

address_customer1 = customer1['address']
print(address_customer1)
````

### Understand this carefully

```python
list_of_dict[0]
```

The list uses an **index** to select the first dictionary.

Then:

```python
customer1['address']
```

The dictionary uses a **key** to select the address.

So:

> **List → index**  
> **Dictionary → key**

### Line-by-line explanation

1. `customer1 = list_of_dict[0]` selects the first dictionary from the list (index `0`) and stores it in `customer1`.
2. `print(customer1)` displays that entire dictionary.
3. `address_customer1 = customer1['address']` looks up the `'address'` key inside `customer1` and stores its value.
4. `print(address_customer1)` displays just that address.

````python
for i in range(len(list_of_dict)):
    print(list_of_dict[i]['address'])
````

### Line-by-line explanation

1. `len(list_of_dict)` returns `3`, the number of dictionaries in the list.
2. `range(len(list_of_dict))` produces the valid indexes `0`, `1`, `2`.
3. `for i in range(len(list_of_dict)):` loops through those indexes.
4. `list_of_dict[i]` selects the dictionary at position `i` using the list index, and `['address']` then selects that dictionary's address using the dictionary key.
5. `print(...)` displays each person's address, one per line.
