## 1. Variables

A **variable** is a name that refers to a value.

Think of it like a label on a box:

```text
name → "Saad"
age  → 17
```

Basic syntax:
```python
variable_name = value
```

Python is dynamically typed, so you do not need to declare the type before assigning a value.

````python
name = "Saad"
age = 17

print(name)
print(age)
````

### Line-by-line explanation

1. `name = "Saad"` creates a variable called `name` and stores the text `"Saad"` in it.
2. `age = 17` creates a variable called `age` and stores the whole number `17` in it.
3. `print(name)` displays the value stored in `name`.
4. `print(age)` displays the value stored in `age`.



*A variable is a label pointing to a value stored in memory.*

## 2. Types of Variables / Values

Python has four basic types covered in this class:

| Type | Example | Meaning |
|---|---|---|
| Integer (`int`) | `17` | Whole number |
| Float (`float`) | `60.17` | Decimal number |
| String (`str`) | `"Saad"` | Text |

Use `type()` to check a value's type.



*Python's four basic data types: str, int, float, and bool.*

3. `weight = 60.17` stores a decimal number, so it is a Float.
4. `is_present = True` stores a true/false value, so it is a Boolean.
5. `print(type(name))` displays `<class 'str'>`.
6. `print(type(age))` displays `<class 'int'>`.
7. `print(type(weight))` displays `<class 'float'>`.
8. `print(type(is_present))` displays `<class 'bool'>`.

## 7. Comments in Python

A **comment** is a note written for humans. Python ignores comments when executing the program.

A single-line comment starts with `#`.

```python
# This is a comment
name = "Saad"  # This is also a comment
```

````python
# Store the student's name
name = "Saad"

# Display the name
print(name)
````

### Line-by-line explanation

1. `# Store the student's name` is a comment. Python skips this line completely; it does not run as code.
2. `name = "Saad"` creates the variable.
3. `# Display the name` is another comment, also skipped by Python.
4. `print(name)` displays the value.
5. Anything written after `#` on a line is a comment. Comments help make code easier to understand for other people (and for you, later).

## 8. Operators

Operators are symbols or keywords used to perform operations on values or variables.

This class covers **7 categories**: Arithmetic, Assignment, Comparison, Logical, Identity, Membership, and Bitwise.

### 8.1 Arithmetic Operators

`+`, `-`, `*`, `/`, `%`, `//`, `**` perform mathematical operations.

````python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.333...
print(a % b)   # 1
print(a ** b)  # 1000
print(a // b)  # 3
````

### Line-by-line explanation

1. `a = 10` and `b = 3` set the two starting numbers.
2. `a + b` adds them, giving `13`.
3. `a - b` subtracts `b` from `a`, giving `7`.
4. `a * b` multiplies them, giving `30`.
5. `a / b` divides them and always returns a Float, giving `3.333...`.
6. `a % b` returns the remainder of the division, giving `1`.
7. `a ** b` raises `a` to the power of `b` (10 to the power 3), giving `1000`.
8. `a // b` performs floor division (division rounded down to the nearest whole number), giving `3`.

### 8.2 Assignment Operators

`=`, `+=`, `-=`, `*=`, `/=`, `%=`, `**=`, `//=` assign or update values.

For example, `a += b` means `a = a + b`.

````python
a = 10
b = 3

a += b; print(a)
a -= b; print(a)
a *= b; print(a)
a /= b; print(a)
a %= b; print(a)
a **= b; print(a)
a //= b; print(a)
````

### Line-by-line explanation

1. `a = 10` and `b = 3` set the starting values.
2. `a += b` means `a = a + b`, so `a` becomes `13`.
3. `a -= b` means `a = a - b`, using the new `a` (13), so `a` becomes `10`.
4. `a *= b` means `a = a * b`, so `a` becomes `30`.
5. `a /= b` means `a = a / b`, so `a` becomes `10.0` (division always returns a Float).
6. `a %= b` means `a = a % b`, so `a` becomes `1.0`.
7. `a **= b` means `a = a ** b`, so `a` becomes `1.0`.
8. `a //= b` means `a = a // b`, so `a` becomes `0.0`.
9. The semicolon `;` lets us write a statement and a `print()` call on the same line, as two separate instructions.

### 8.3 Comparison Operators

Comparison operators compare values and return `True` or `False`:

`==`, `!=`, `>`, `<`, `>=`, `<=`.

````python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
````

### Line-by-line explanation

1. `a = 10` and `b = 5` set the two numbers being compared.
2. `a == b` checks whether they are equal. `10` is not `5`, so this is `False`.
3. `a != b` checks whether they are **not** equal, so this is `True`.
4. `a > b` checks whether `a` is greater than `b`, so this is `True`.
5. `a < b` checks whether `a` is less than `b`, so this is `False`.
6. `a >= b` checks greater-than-or-equal, so this is `True`.
7. `a <= b` checks less-than-or-equal, so this is `False`.
8. Remember: `=` **assigns** a value, while `==` **compares** two values.

### 8.4 Logical Operators

- `and` → both conditions must be `True`
- `or` → at least one condition must be `True`
- `not` → reverses the result

````python
x = True
y = False

print(x and y)
print(x or y)
print(not x)
````

### Line-by-line explanation

1. `x = True` and `y = False` set the two Boolean values.
2. `x and y` needs **both** to be `True`. Since `y` is `False`, the result is `False`.
3. `x or y` only needs **one** of them to be `True`. Since `x` is `True`, the result is `True`.
4. `not x` reverses `x`. Since `x` is `True`, the result is `False`.

### 8.5 Identity Operators

`is` and `is not` check whether two variables refer to the **same object** in memory (not just equal values).

````python
a = [1, 2, 3]
b = a

print(a is b)
print(a is not b)
````

### Line-by-line explanation

1. `a = [1, 2, 3]` creates a list and stores it in `a`.
2. `b = a` does **not** create a new list. `b` now points to the exact same list object as `a`.
3. `a is b` checks whether both names point to the same object in memory. Since they do, this is `True`.
4. `a is not b` checks the opposite, so this is `False`.

### 8.6 Membership Operators

`in` and `not in` check whether a value exists inside a sequence such as a list or String.

````python
nums = [1, 2, 3]

print(2 in nums)
print(5 not in nums)
````

### Line-by-line explanation

1. `nums = [1, 2, 3]` creates a list of numbers.
2. `2 in nums` checks whether `2` exists inside `nums`. Since it does, this is `True`.
3. `5 not in nums` checks whether `5` does **not** exist inside `nums`. Since `5` is missing, this is `True`.

### 8.7 Bitwise Operators

Bitwise operators work with the binary representation of integers: `&`, `|`, `^`, `~`, `<<`, `>>`.

````python
a = 5    # 101
b = 3    # 011

print(a & b)
print(a | b)
print(a ^ b)
print(a << 1)
print(a >> 1)
````

### Line-by-line explanation

1. `a = 5` is `101` in binary, and `b = 3` is `011` in binary.
2. `a & b` (AND) keeps a `1` only where **both** bits are `1`. That gives `001`, which is `1`.
3. `a | b` (OR) gives a `1` where **at least one** bit is `1`. That gives `111`, which is `7`.
4. `a ^ b` (XOR) gives a `1` where the bits are **different**. That gives `110`, which is `6`.
5. `a << 1` shifts the bits of `a` one place to the left, which doubles the value, giving `10`.
6. `a >> 1` shifts the bits of `a` one place to the right, which halves the value (rounded down), giving `2`.

## 10. Concatenation

**Concatenation** means joining things together.

For Strings, `+` can join text:

```python
"Saad" + "Ahmed"
```

A comma in `print()` is different: it prints multiple values separately, with a space automatically added between them.

````python
print("Saad" + "Ahmed")
print("Saad", "Ahmed")
````

### Line-by-line explanation

1. `print("Saad" + "Ahmed")` joins the two Strings directly with `+`, so it prints `SaadAhmed` with no space in between.
2. `print("Saad", "Ahmed")` prints both values separated by a single space, because a comma inside `print()` adds spacing automatically between each value.

### Method 1 — Commas in `print()`

`print()` can display multiple values separated by commas. This is useful when values have different types, because `print()` handles the conversion to text for you.

````python
name = "Saad"
age = 17
weight = 60.1
is_present = True

print("Student Name:", name)
print("Student Age:", age)
print("Student Weight:", weight)
print("Student is Present:", is_present)
````

### Line-by-line explanation

1. Four variables are created, storing a String, an Integer, a Float, and a Boolean.
2. `print("Student Name:", name)` passes a label String and the variable, separated by a comma.
3. Python prints them with a space in between; it does **not** need `str()` here because commas can handle any type automatically.
4. The remaining three `print()` calls repeat this pattern for `age`, `weight`, and `is_present`.

### Method 2 — `+`

When using `+` to join text, **every value must already be a String**. Use `str()` to convert non-String values first.

````python
name = "Saad"
age = 17
weight = 60.1
is_present = True

print("Student Name: " + name)
print("Student Age: " + str(age))
print("Student Weight: " + str(weight))
print("Student is Present: " + str(is_present))
````

### Line-by-line explanation

1. `"Student Name: " + name` works directly with `+` because both sides are already Strings.
2. `"Student Age: " + str(age)` needs `str(age)` because `age` is an Integer; `+` cannot join text with a number directly, so `str()` converts the number into text first.
3. The same reasoning applies to `weight` (a Float) and `is_present` (a Boolean) — both are converted with `str()` before joining.
4. Without `str()`, writing `"Student Age: " + age` would raise a `TypeError`, because Python refuses to add a String and an Integer with `+`.

### Method 3 — f-strings

An f-string lets you put variables directly inside `{}`.

````python
name = "Saad"
age = 17
weight = 60.1
is_present = True

print(f"Student Name: {name}")
print(f"Student Age: {age}")
print(f"Student Weight: {weight}")
print(f"Student is Present: {is_present}")
````

### Line-by-line explanation

1. The letter `f` before the opening quote tells Python this is an f-string.
2. Anything written inside `{}` is treated as a variable (or expression), and its value is inserted directly into the text.
3. Unlike Method 2, f-strings do **not** need `str()` — Python automatically converts every value inside `{}` to text, no matter its type.
4. This makes f-strings the most convenient of the three methods for most beginners.

### More examples

````python
print("The sum of 2 plus 2 is", 2 + 2)
print("The sum of 2 plus 2 is " + "4")
````

### Line-by-line explanation

1. `print("The sum of 2 plus 2 is", 2 + 2)` uses a comma, so Python first calculates `2 + 2` (giving `4`) and then prints the label and the number `4`, separated by a space.
2. `print("The sum of 2 plus 2 is " + "4")` uses `+`, so both sides must already be Strings. Here `"4"` is written directly as text, so no conversion is needed — but note this is text `"4"`, not the calculated number.

````python
greeting = "Hello"
separator = ", "
addressee = "World"
punctuation = "!"

whole_greeting = greeting + separator + addressee + punctuation
print(whole_greeting)
````

### Line-by-line explanation

1. Four small String pieces are created: `greeting`, `separator`, `addressee`, and `punctuation`.
2. `greeting + separator + addressee + punctuation` joins all four Strings together, in order, into one combined String.
3. The result, `"Hello, World!"`, is stored in `whole_greeting`.
4. `print(whole_greeting)` displays the final joined String.

````python
f_name = "Saad"
l_name = "Ahmed"
age = 17
weight = 60.12
is_trainer = True

print("Trainer Name of Python class is " + f_name + " " + l_name)
print(f"Trainer Name of Python class is {f_name} {l_name}")
print("Trainer Name of Python class is", f_name, l_name)
print("Trainer Name of Python class is {0} {1}".format(f_name, l_name))
````

### Line-by-line explanation

1. The first `print()` uses `+` to join Strings directly; a separate `" "` String is added between `f_name` and `l_name` so the names do not run together.
2. The second `print()` uses an f-string, inserting `{f_name}` and `{l_name}` directly, with a normal space typed between the two braces.
3. The third `print()` uses commas, letting `print()` insert the spacing automatically between each value.
4. The fourth `print()` uses `.format()`, where `{0}` and `{1}` are placeholders that refer to the first and second arguments passed to `.format()` — `{0}` becomes `f_name` and `{1}` becomes `l_name`.
5. All four lines print the same sentence; they are simply four different ways to combine text and variables.
