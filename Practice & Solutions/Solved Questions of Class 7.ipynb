## Dictionary Practice

### Question 1
Create a dictionary containing your name, city, and favorite subject. Use `keys()`, `values()`, and `items()` on it.

````python
my_dict = {
    "name": "Saad",
    "city": "Karachi",
    "favorite_subject": "Computer Science"
}

print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())
````

**Line-by-line explanation**
- `my_dict = {...}` → creates a dictionary with three key-value pairs.
- `my_dict.keys()` → returns a **view** of all the keys: `dict_keys(['name', 'city', 'favorite_subject'])`.
- `my_dict.values()` → returns a view of all the values: `dict_values(['Saad', 'Karachi', 'Computer Science'])`.
- `my_dict.items()` → returns a view of every key paired with its value, as tuples: `dict_items([('name', 'Saad'), ('city', 'Karachi'), ('favorite_subject', 'Computer Science')])`.

### Question 2
Why can't `my_dict.items()[0]` be used directly? Explain what `items()` returns.

**Answer (conceptual)**

`my_dict.items()` doesn't return a list or a tuple — it returns a special object called a **`dict_items` view**. A view object supports looping (`for key, value in my_dict.items():`) and checking membership, but it does **not** support indexing with `[0]`, because a dictionary itself has no guaranteed positional order for square-bracket access the way a list does.

Trying `my_dict.items()[0]` raises:
```text
TypeError: 'dict_items' object is not subscriptable
```

To get the first item by position, you first need to **convert it into a list**:
```python
list(my_dict.items())[0]
```
This is exactly what Question 3 asks you to do.

### Question 3
Convert a dictionary's items into a list and access the first key-value pair.

````python
my_dict = {
    "name": "Saad",
    "city": "Karachi",
    "favorite_subject": "Computer Science"
}

items_list = list(my_dict.items())
first_item = items_list[0]
print(first_item)
````

**Line-by-line explanation**
- `my_dict.items()` → returns the `dict_items` view (not indexable, as explained in Question 2).
- `list(my_dict.items())` → converts that view into an actual **list** of tuples, which **is** indexable.
- `items_list[0]` → now that it's a plain list, standard list indexing works, giving us the first tuple: `('name', 'Saad')`.

### Question 4
Write a `for` loop that prints every key and value separately.

````python
my_dict = {
    "name": "Saad",
    "city": "Karachi",
    "favorite_subject": "Computer Science"
}

for key, value in my_dict.items():
    print(key)
    print(value)
````

**Line-by-line explanation**
- `for key, value in my_dict.items():` → unpacks each key-value tuple into two separate variables at once.
- Inside the loop, `print(key)` and `print(value)` are two **separate** statements — printing the key and the value on their own individual lines, rather than together on one line.

### Question 5
Explain the difference between a dictionary key, value, and item.

**Answer (conceptual)**

- A **key** is the unique label used to identify a piece of data — e.g. `"name"`. Keys must be unique within a dictionary.
- A **value** is the actual data stored under that label — e.g. `"Saad"`. Values can repeat, and can be of any type.
- An **item** is the key and its value **together**, as a pair — e.g. `("name", "Saad")`. `.items()` is what gives you both halves at once, which is why it's the standard way to loop through a dictionary when you need both the key and its value.

## Lists of Dictionaries Practice

### Question 1
Create a list containing three student dictionaries.

````python
students = [
    {"name": "Ali", "grade": "A"},
    {"name": "Ahmed", "grade": "B"},
    {"name": "Saad", "grade": "A"}
]

print(students)
````

**Line-by-line explanation**
- `students = [...]` → a list where each item is a dictionary representing one student's record.

### Question 2
Access the last student's name using a negative index.

````python
students = [
    {"name": "Ali", "grade": "A"},
    {"name": "Ahmed", "grade": "B"},
    {"name": "Saad", "grade": "A"}
]

print(students[-1]["name"])
````

**Line-by-line explanation**
- `students[-1]` → negative indexing gets the **last** dictionary in the list, regardless of how many students there are.
- `["name"]` → then looks up the `"name"` key inside that dictionary, printing `"Saad"`.

### Question 3
Loop through the list and print every student's name.

````python
students = [
    {"name": "Ali", "grade": "A"},
    {"name": "Ahmed", "grade": "B"},
    {"name": "Saad", "grade": "A"}
]

for student in students:
    print(student["name"])
````

**Line-by-line explanation**
- `for student in students:` → each time through the loop, `student` holds one **entire dictionary** from the list.
- `student["name"]` → pulls out just the `"name"` key from whichever dictionary is currently being visited, printing each student's name in turn.

### Question 4
Add a new student dictionary using `append()`.

````python
students = [
    {"name": "Ali", "grade": "A"},
    {"name": "Ahmed", "grade": "B"},
    {"name": "Saad", "grade": "A"}
]

students.append({"name": "Hina", "grade": "A"})
print(students)
````

**Line-by-line explanation**
- `students.append({...})` → `.append()` adds one new item to the end of the list — here, that item happens to be a whole new dictionary.
- After this runs, `students` contains four student records instead of three.

### Question 5
Count how many student records are in the list using `len()`.

````python
students = [
    {"name": "Ali", "grade": "A"},
    {"name": "Ahmed", "grade": "B"},
    {"name": "Saad", "grade": "A"},
    {"name": "Hina", "grade": "A"}
]

print(len(students))
````

**Line-by-line explanation**
- `len(students)` → counts how many **items** are in the list. Since each item is one student dictionary, this directly gives the number of student records — `4`.
- Note that `len()` counts the outer list's items; it does not look inside each dictionary or count keys.

## Nested Dictionaries Practice

### Question 1
Create a nested dictionary containing two courses and their instructors.

````python
courses = {
    "Python": {"instructor": "Saad", "duration_weeks": 8},
    "Web Development": {"instructor": "Ahmed", "duration_weeks": 10}
}

print(courses)
````

**Line-by-line explanation**
- `courses = {...}` → the **outer** dictionary has two keys, `"Python"` and `"Web Development"`.
- Each of those keys maps to its own **inner** dictionary, holding details specific to that course (`instructor`, `duration_weeks`). This "dictionary inside a dictionary" structure is what makes it **nested**.

### Question 2
Access the instructor of one course.

````python
courses = {
    "Python": {"instructor": "Saad", "duration_weeks": 8},
    "Web Development": {"instructor": "Ahmed", "duration_weeks": 10}
}

print(courses["Python"]["instructor"])
````

**Line-by-line explanation**
- `courses["Python"]` → looks up the outer key `"Python"`, which returns the **inner dictionary** `{"instructor": "Saad", "duration_weeks": 8}`.
- `["instructor"]` → is then applied directly to that inner dictionary, retrieving `"Saad"`.
- Reading it left to right: go into the `courses` dictionary, find `Python`, then inside that, find `instructor`.

### Question 3
Loop through the nested dictionary using `.items()`.

````python
courses = {
    "Python": {"instructor": "Saad", "duration_weeks": 8},
    "Web Development": {"instructor": "Ahmed", "duration_weeks": 10}
}

for course_name, course_info in courses.items():
    print(course_name, "-->", course_info)
````

**Line-by-line explanation**
- `courses.items()` → gives each **outer** key-value pair together — here, `course_name` is the outer key (e.g. `"Python"`), and `course_info` is its entire inner dictionary.
- `print(course_name, "-->", course_info)` → prints the course name next to its full details dictionary, e.g. `Python --> {'instructor': 'Saad', 'duration_weeks': 8}`.

### Question 4
Explain the difference between accessing a value through an outer key and through an inner dictionary.

**Answer (conceptual)**

- Accessing through the **outer key** (e.g. `courses["Python"]`) gives you the **entire inner dictionary** for that course — everything nested underneath it, as one object.
- Accessing through the **inner dictionary's key** (e.g. `courses["Python"]["instructor"]`) drills one level deeper, past the outer key, and retrieves a **single specific value** stored inside that inner dictionary.

In short: one outer key access gets you the whole record, while chaining a second key access on top of it gets you one specific field within that record.

### Question 5
Create a nested dictionary for three products containing price and quantity.

````python
products = {
    "Laptop": {"price": 150000, "quantity": 5},
    "Mouse": {"price": 1500, "quantity": 20},
    "Keyboard": {"price": 3000, "quantity": 15}
}

for product_name, product_info in products.items():
    print(product_name, "-->", product_info)
````

**Line-by-line explanation**
- `products = {...}` → the outer dictionary has three keys (product names), each mapping to its own inner dictionary containing `price` and `quantity`.
- The loop then displays every product together with its full details, using the exact same `.items()` pattern from Question 3.

## Functions Practice

### Question 1
Create a function that prints a short welcome message and call it twice.

````python
def print_welcome():
    print("Welcome to Python Class 7!")

print_welcome()
print_welcome()
````

**Line-by-line explanation**
- `def print_welcome():` → defines a new function named `print_welcome`. The code inside only runs when the function is **called**, not when it's defined.
- `print("Welcome to Python Class 7!")` → the single line of code inside the function's body.
- `print_welcome()` (called twice) → each call runs the function's body from scratch, printing the welcome message once per call — so the message appears twice in total.

### Question 2
Create a function that accepts two numbers and prints their product.

````python
def multiply(num1, num2):
    print(num1 * num2)

multiply(4, 5)
````

**Line-by-line explanation**
- `def multiply(num1, num2):` → defines a function with two **parameters**, `num1` and `num2` — placeholders for whatever values get passed in later.
- `print(num1 * num2)` → multiplies the two received values and prints the result.
- `multiply(4, 5)` → calls the function with `4` and `5` as **arguments**; `num1` becomes `4` and `num2` becomes `5`, so this prints `20`.

### Question 3
Create a function with one required parameter and one default parameter.

````python
def greet(name, greeting="Hello"):
    print(greeting, name)

greet("Saad")
greet("Arfa", "Hi")
````

**Line-by-line explanation**
- `def greet(name, greeting="Hello"):` → `name` is a **required** parameter (no default given, so it must always be supplied). `greeting="Hello"` is an **optional** parameter — if no value is passed for it, it automatically becomes `"Hello"`.
- `greet("Saad")` → only `name` is supplied, so `greeting` falls back to its default, `"Hello"`, printing `Hello Saad`.
- `greet("Ahmed", "Hi")` → this time a second value is provided, **overriding** the default, printing `Hi Ahmed`.

### Question 4
Write one function call using positional arguments and another using keyword arguments.

````python
def introduce(first_name, last_name):
    print(first_name, last_name)

# Positional arguments: matched by order
introduce("Saad", "Ahmed")

# Keyword arguments: matched by name, order doesn't matter
introduce(last_name="Ahmed", first_name="Saad")
````

**Line-by-line explanation**
- `introduce("Saad", "Ahmed")` → **positional** arguments — Python matches them to parameters purely by their **order**: the 1st value goes to `first_name`, the 2nd to `last_name`.
- `introduce(last_name="Ahmed", first_name="Saad")` → **keyword** arguments — each value is explicitly labeled with the parameter name it belongs to, so the **order no longer matters**. Both calls produce the exact same output, `Saad Ahmed`, because the values still end up matched to the correct parameters either way.

### Question 5
Create a function that returns a calculated value, store that returned value in a variable, and print it.

````python
def calculate_area(length, width):
    return length * width

area = calculate_area(5, 3)
print(area)
````

**Line-by-line explanation**
- `return length * width` → unlike `print()`, `return` sends the calculated value **back out** of the function so it can be used elsewhere in the program, rather than just displaying it.
- `area = calculate_area(5, 3)` → calls the function, and because it uses `return`, the result (`15`) is captured and stored in the variable `area`.
- `print(area)` → the value is printed afterward, from the variable — demonstrating that a returned value can be saved and used later, unlike a printed value which only appears on screen and can't be reused.
