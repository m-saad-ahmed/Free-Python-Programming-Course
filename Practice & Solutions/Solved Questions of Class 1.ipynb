# Section 1 — Practice: What is Python?

### Question 1 — What is Python?

**Answer:**

Python is a **high-level, interpreted, general-purpose programming language**. "High-level" means it is written in a way that is close to human language rather than machine code, so it is easier for people to read and write. It lets us give a computer a precise set of instructions (a *program*) to perform a task.

### Question 2 — Who created Python?

**Answer:**

Python was created by **Guido van Rossum**.

### Question 3 — When was Python first released?

**Answer:**

Python was first released in **1991**.

### Question 4 — Why is Python considered beginner-friendly?

**Answer:**

Python is considered beginner-friendly because its **syntax is simple and reads almost like plain English** (for example, `print("Hello")` clearly shows its intent). It doesn't require complicated setup for small programs, has helpful error messages, and has a huge amount of learning material and community support available.

### Question 5 — Name two things you can build or do with Python.

**Answer:**

Any two of the following are correct: websites and web applications (web development), data analysis, artificial intelligence / machine learning projects, automating repetitive tasks, games, or cloud/DevOps tooling.

# Section 3 — Practice: Features of Python

### Question 1 — What does "easy to learn" mean for Python?

**Answer:**

It means Python's syntax (the rules for writing code) is simple, clean, and close to plain English, so beginners can start writing working programs quickly without memorizing a lot of complicated rules.

### Question 2 — What does interpreted mean?

**Answer:**

"Interpreted" means Python code is run **line by line by a program called the interpreter**, rather than being fully translated into machine code ahead of time (as "compiled" languages like C do). This is why you can run a single Python line immediately and see its result.

### Question 3 — What does object-oriented mean?

**Answer:**

"Object-oriented" means Python supports organizing code around **classes and objects** — bundling related data and behavior together (this is covered in depth in Class 8 of this course).

### Question 4 — What does portable mean?

**Answer:**

"Portable" means a Python program written on one operating system (say, Windows) can generally run on another (macOS or Linux) with little or no change, because the Python interpreter itself handles the differences between operating systems.

### Question 5 — Why are libraries useful in programming?

**Answer:**

Libraries are useful because they contain **ready-made, tested code** for common tasks, so a programmer doesn't have to write everything from scratch. For example, instead of writing your own code to work with tables of data, you can `import pandas` (Class 11) and use code that thousands of developers have already built and tested.

### Question 4 — Write a program that prints three separate lines of text.

````python
print("Line one")
print("Line two")
print("Line three")
````

**Line-by-line explanation**

1. `print("Line one")` displays the first line, then automatically moves the cursor to a new line afterward — `print()` always adds a newline at the end by default.
2. `print("Line two")` displays the second line, starting fresh on its own row because of that automatic newline from the previous `print()`.
3. `print("Line three")` displays the third line, the same way.
4. Because each `print()` call is a separate statement, the output naturally appears as three separate lines, without needing anything extra.

### Question 5 — Explain in your own words what `print()` does.

**Answer:**

`print()` is a built-in Python function that **displays information on the screen** (in the console or notebook output). Whatever value is placed inside its parentheses — text, a number, or a variable — gets shown to whoever is running the program. It is one of the most basic and most frequently used tools for seeing what a program is doing.
