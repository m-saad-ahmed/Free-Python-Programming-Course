````python
message = """This is line one.
This is line two.
This is line three."""

print(message)
````

1. `"""..."""` (triple double quotes) lets a String span multiple lines exactly as typed, including the line breaks inside it.

### Question 5 — Write a String containing quotation marks inside it.

````python
quote = "She said, \"Python is fun!\""
print(quote)

quote2 = 'He said, "Let\'s learn Python."'
print(quote2)
````

**Line-by-line explanation**
1. `quote` is written with double quotes as the outer boundary; the double quotes that appear *inside* the text are "escaped" with a backslash (`\\"`) so Python knows they are part of the text, not the end of the String.
2. `print(quote)` displays: `She said, "Python is fun!"`
3. `quote2` shows an alternative approach: using **single** quotes as the outer boundary means the double quotes inside don't need escaping at all — only the apostrophe (`'`) inside needs a backslash (`\\'`).
4. This demonstrates a useful trick: choosing single vs. double quotes as your outer boundary can avoid needing to escape most inner quotation marks.

# Section 12 — Practice: Identity

### Question 1 — Create a list called `a`.

````python
a = [1, 2, 3]
print(a)
````

**Line-by-line explanation**
1. `a = [1, 2, 3]` creates a list (Class 4) with three numbers.
2. `print(a)` displays `[1, 2, 3]`.

### Question 2 — Assign `b = a` and test `a is b`.

````python
a = [1, 2, 3]
b = a

print(a is b)
````

**Line-by-line explanation**
1. `b = a` does **not** create a new list — `b` now points to the exact same list object as `a` in memory.
2. `a is b` checks whether both names refer to the same object. Since they do, the result is `True`.

### Question 3 — Create another list with the same values and compare it using `is`.

````python
a = [1, 2, 3]
c = [1, 2, 3]

print(a == c)
print(a is c)
````

**Line-by-line explanation**
1. `c = [1, 2, 3]` creates a **brand-new, separate** list that happens to contain the same values as `a`.
2. `a == c` → `True`, because `==` compares the *values* inside the lists, and they match.
3. `a is c` → `False`, because `is` compares *identity* (are they literally the same object in memory?) — and they are not, since `c` was built independently.

### Question 4 — Test `is not` with two variables.

````python
a = [1, 2, 3]
c = [1, 2, 3]

print(a is not c)
````

**Line-by-line explanation**
1. `a is not c` checks the opposite of `is`. Since `a` and `c` are different objects (as shown in Question 3), the result is `True`.

### Question 5 — Explain in your own words what `is` checks.

**Answer:**

`is` checks whether two variables point to the **exact same object in memory** — not just whether their values look the same. Two separate lists can contain identical values (so `==` says `True`) while still being two different objects in memory (so `is` says `False`).

# Section 14 — Practice: Bitwise

### Question 1 — Calculate `5 & 3`.

````python
print(5 & 3)
````

**Line-by-line explanation**
1. `5` is `101` in binary, `3` is `011`. `&` (AND) keeps a `1` only where **both** bits are `1`: `001` → `1`.

### Question 2 — Calculate `5 | 3`.

````python
print(5 | 3)
````

**Line-by-line explanation**
1. `|` (OR) keeps a `1` where **at least one** bit is `1`: `101 | 011 = 111` → `7`.

### Question 3 — Calculate `5 ^ 3`.

````python
print(5 ^ 3)
````

**Line-by-line explanation**
1. `^` (XOR) keeps a `1` where the bits are **different**: `101 ^ 011 = 110` → `6`.

### Question 4 — Calculate `5 << 1`.

````python
print(5 << 1)
````

**Line-by-line explanation**
1. `<<` shifts bits left by 1 position, which doubles the value: `5 << 1 = 10`.

### Question 5 — Calculate `5 >> 1`.

````python
print(5 >> 1)
````

**Line-by-line explanation**
1. `>>` shifts bits right by 1 position, which halves the value (rounded down): `5 >> 1 = 2`.

### Question 2 — Is `if` a keyword?

````python
import keyword
print(keyword.iskeyword("if"))
````

**Line-by-line explanation**
1. `import keyword` brings in Python's built-in `keyword` module, which knows every reserved word.
2. `keyword.iskeyword("if")` checks whether `"if"` is a reserved keyword, and returns `True` — confirming `if` is indeed a keyword.

### Question 3 — Why can't keywords normally be used as ordinary variable names?

**Answer:**

Keywords are **reserved** by Python because they have a special, fixed meaning in the language's own grammar (for example, `if` always starts a conditional statement). If Python allowed `if = 5`, it would become impossible for Python to tell whether `if` means "the start of a condition" or "a variable I just created" — so keywords are blocked from being used as ordinary names to avoid this ambiguity.

### Question 4 — Identify the keywords in a statement containing `if` and `else`.

````python
age = 20

if age >= 18:
    print("Adult")
else:
    print("Minor")
````

**Line-by-line explanation**
1. In this statement, `if` and `else` are the two **keywords** — reserved words that control which branch of code runs.
2. `age`, `print` are not keywords (`age` is a normal variable name; `print` is a built-in function name, which is different from a keyword — it *can* technically be reassigned, though doing so is bad practice).
3. The condition `age >= 18` is `True` (since `age` is `20`), so `"Adult"` is printed.

3. `age` is an Integer, so it must be wrapped in `str(age)` first to convert it to text before `+` can join it — without `str()`, this would raise a `TypeError`.

### Question 4 — Print a name and age using an f-string.

````python
name = "Saad"
age = 18

print(f"Name: {name} Age: {age}")
````

1. The `f` before the opening quote marks this as an f-string.
2. `{name}` and `{age}` are automatically replaced with the current values of those variables — no `str()` conversion needed, since f-strings handle any type automatically.
3. This displays the same sentence: `Name: Saad Age: 25`.

### Question 5 — Create a greeting using `.format()`.
