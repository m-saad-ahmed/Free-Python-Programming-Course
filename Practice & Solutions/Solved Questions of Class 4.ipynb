## Topic 1 — Creating and Indexing Lists

### Question 1
Create a list named `fruits` containing five fruit names and print the list.

````python
fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]
print(fruits)
````

**Line-by-line explanation**
- `fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]` → creates a list with five string items and stores it in the variable `fruits`.
- `print(fruits)` → prints the entire list, including the square brackets and commas that Python uses to display a list.

### Question 2
Print the first item from the list.

````python
fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]
print(fruits[0])
````

**Line-by-line explanation**
- `fruits[0]` → accesses the item at index `0`. Python lists are **zero-indexed**, so the first item is always at position `0`.
- `print(...)` → displays that value, `Apple`.

### Question 3
Print the third item from the list.

````python
fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]
print(fruits[2])
````

**Line-by-line explanation**
- `fruits[2]` → the third item is at index `2`, not `3`, because counting starts from `0` (index `0` = 1st item, index `1` = 2nd item, index `2` = 3rd item).
- This prints `Mango`.

### Question 4
Print the last item using a negative index.

````python
fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]
print(fruits[-1])
````

**Line-by-line explanation**
- `fruits[-1]` → negative indexing counts from the end of the list. `-1` always refers to the **last** item, regardless of the list's length.
- This prints `Orange`.

### Question 5
Print the second-last item using a negative index.

````python
fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]
print(fruits[-2])
````

**Line-by-line explanation**
- `fruits[-2]` → `-1` is the last item, so `-2` is the item right before it — the second-last item.
- This prints `Grapes`.

## Topic 2 — Changing and Slicing

### Question 1
Create a list of six countries and change the second country.

````python
countries = ["Pakistan", "India", "China", "Japan", "France", "Brazil"]
countries[1] = "Germany"
print(countries)
````

**Line-by-line explanation**
- `countries = [...]` → creates a list of six country names.
- `countries[1] = "Germany"` → since lists are **mutable**, we can reassign the value at any index. Index `1` is the second item (`"India"`), which gets replaced by `"Germany"`.
- `print(countries)` → shows the updated list, with `"India"` now gone and `"Germany"` in its place.

### Question 2
Print the first four countries using slicing.

````python
countries = ["Pakistan", "India", "China", "Japan", "France", "Brazil"]
print(countries[:4])
````

**Line-by-line explanation**
- `countries[:4]` → slicing syntax is `list[start:stop]`. When `start` is left empty, Python assumes `0` (the beginning).
- `4` as the stop value means "up to, but **not including**, index 4" — so this returns indices `0, 1, 2, 3`, which are the first four items.

### Question 3
Print the middle three countries using slicing.

````python
countries = ["Pakistan", "India", "China", "Japan", "France", "Brazil"]
print(countries[2:5])
````

**Line-by-line explanation**
- The list has 6 items (indices `0` to `5`). The three "middle" items are at indices `2`, `3`, and `4` (`China`, `Japan`, `France`).
- `countries[2:5]` → starts at index `2` and stops **before** index `5`, giving exactly those three middle items.

### Question 4
Print the last three countries using a negative slice.

````python
countries = ["Pakistan", "India", "China", "Japan", "France", "Brazil"]
print(countries[-3:])
````

**Line-by-line explanation**
- `countries[-3:]` → starting from `-3` (three items counted back from the end) and going all the way to the end (empty `stop` means "go to the end of the list").
- This returns the last three items: `Japan`, `France`, `Brazil`.

### Question 5
Print the list in reverse order using slicing.

````python
countries = ["Pakistan", "India", "China", "Japan", "France", "Brazil"]
print(countries[::-1])
````

**Line-by-line explanation**
- The full slicing syntax is `list[start:stop:step]`. Here, `start` and `stop` are both left empty (meaning "the whole list"), and `step` is `-1`.
- A step of `-1` tells Python to move through the list **backwards**, one item at a time — producing the entire list in reverse order.

## Topic 3 — Adding Items

### Question 1
Create a list of three programming languages and add one using `append()`.

````python
languages = ["Python", "JavaScript", "C++"]
languages.append("Java")
print(languages)
````

**Line-by-line explanation**
- `languages = [...]` → creates a list of three language names.
- `languages.append("Java")` → `.append()` adds exactly **one item** to the **end** of the list. The list is modified in place — no new list is created.
- `print(languages)` → shows `"Java"` now added as the fourth item.

### Question 2
Insert a language at index `1`.

````python
languages = ["Python", "JavaScript", "C++", "Java"]
languages.insert(1, "Ruby")
print(languages)
````

**Line-by-line explanation**
- `languages.insert(1, "Ruby")` → `.insert(index, value)` places `"Ruby"` at index `1`, pushing every item from that position onward one step to the right.
- Unlike `.append()`, `.insert()` lets you choose exactly **where** the new item goes, not just the end.

### Question 3
Create another list containing two languages.

````python
more_languages = ["Go", "Rust"]
print(more_languages)
````

**Line-by-line explanation**
- `more_languages = ["Go", "Rust"]` → creates a completely separate, second list containing two items. This will be used in the next question.

### Question 4
Add both languages using `extend()`.

````python
languages = ["Python", "JavaScript", "C++", "Java"]
more_languages = ["Go", "Rust"]

languages.extend(more_languages)
print(languages)
````

**Line-by-line explanation**
- `languages.extend(more_languages)` → `.extend()` takes every item **inside** `more_languages` and adds them individually to the end of `languages`.
- This is different from `.append(more_languages)`, which would have added the whole list as a **single nested item** instead of merging the two lists.

### Question 5
Combine two lists using `+` and store the result in a new variable.

````python
languages = ["Python", "JavaScript", "C++", "Java"]
more_languages = ["Go", "Rust"]

combined = languages + more_languages
print(combined)
````

**Line-by-line explanation**
- `languages + more_languages` → the `+` operator on two lists creates and returns a **brand-new list** containing all items from both, in order. Neither original list is changed.
- `combined = ...` → stores this new list in a new variable so we can use it separately from `languages` and `more_languages`.

## Topic 4 — Removing Items

### Question 1
Create a list of five foods and remove one using `remove()`.

````python
foods = ["Rice", "Bread", "Pasta", "Salad", "Soup"]
foods.remove("Pasta")
print(foods)
````

**Line-by-line explanation**
- `foods = [...]` → creates a list of five food items.
- `foods.remove("Pasta")` → `.remove(value)` searches the list for the **first occurrence** of that exact value and deletes it. If the value doesn't exist, Python raises a `ValueError`.
- Note that `.remove()` takes the **value**, not the index.

### Question 2
Delete the first item using `del`.

````python
foods = ["Rice", "Bread", "Salad", "Soup"]
del foods[0]
print(foods)
````

**Line-by-line explanation**
- `del foods[0]` → `del` is a Python keyword (not a list method) that deletes the item at the given index directly from the list.
- Here it removes `"Rice"`, the item at index `0`.

### Question 3
Remove the last item using `pop()`.

````python
foods = ["Bread", "Salad", "Soup"]
foods.pop()
print(foods)
````

**Line-by-line explanation**
- `foods.pop()` → when called with **no argument**, `.pop()` removes and returns the **last** item in the list.
- Here, `"Soup"` is removed, and the list is updated in place.

### Question 4
Remove an item from a specific index using `pop(index)`.

````python
foods = ["Bread", "Salad"]
foods.pop(0)
print(foods)
````

**Line-by-line explanation**
- `foods.pop(0)` → passing an index tells `.pop()` to remove the item at **that specific position** instead of the last one.
- Here, index `0` (`"Bread"`) is removed, leaving only `"Salad"`.

### Question 5
Store the value returned by `pop()` in a variable and print it.

````python
foods = ["Bread", "Salad", "Soup"]
removed_food = foods.pop()
print(removed_food)
print(foods)
````

**Line-by-line explanation**
- `removed_food = foods.pop()` → unlike `.remove()` and `del`, `.pop()` **returns** the item it removes. Here we capture that returned value in the variable `removed_food`.
- `print(removed_food)` → shows the item that was taken out (`"Soup"`).
- `print(foods)` → shows the list afterward, confirming that `"Soup"` is no longer in it.

## Topic 5 — Useful List Methods

### Question 1
Find the length of a list containing seven numbers.

````python
numbers = [4, 8, 15, 16, 23, 42, 7]
print(len(numbers))
````

**Line-by-line explanation**
- `numbers = [...]` → a list of seven integers.
- `len(numbers)` → `len()` is a built-in Python function that counts how many items are in the list. It returns `7`.

### Question 2
Check whether a particular value exists using `in`.

````python
numbers = [4, 8, 15, 16, 23, 42, 7]
print(15 in numbers)
print(100 in numbers)
````

**Line-by-line explanation**
- `15 in numbers` → the `in` keyword checks whether a value exists anywhere in the list, and returns a **Boolean** (`True` or `False`). Since `15` is present, this returns `True`.
- `100 in numbers` → `100` is not in the list, so this returns `False`.

### Question 3
Count how many times a value appears.

````python
numbers = [4, 8, 4, 15, 4, 16]
print(numbers.count(4))
````

**Line-by-line explanation**
- `numbers.count(4)` → `.count(value)` scans the entire list and returns how many times that exact value appears.
- Here, `4` appears three times, so the output is `3`.

### Question 4
Find the first index of a value.

````python
numbers = [4, 8, 4, 15, 4, 16]
print(numbers.index(4))
````

**Line-by-line explanation**
- `numbers.index(4)` → `.index(value)` returns the position of the **first** time that value appears in the list. Even though `4` appears three times, this only returns the position of the first one.
- Since `4` is at index `0`, the output is `0`.

### Question 5
Sort a list of numbers and then reverse it.

````python
numbers = [4, 8, 15, 16, 23, 42, 7]

numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)
````

**Line-by-line explanation**
- `numbers.sort()` → sorts the list **in place**, in ascending order by default. The original list itself is changed; nothing is returned.
- `print(numbers)` → shows the list now sorted from smallest to largest.
- `numbers.reverse()` → reverses the current order of the list in place — since it's already sorted ascending, this now puts it in descending order.
- `print(numbers)` → shows the final result, sorted from largest to smallest.

## Topic 6 — Assignment and Copying

### Question 1
Create a list `a`, assign `b = a`, change `b`, and observe what happens to `a`.

````python
a = [1, 2, 3]
b = a

b[0] = 99

print(a)
print(b)
````

**Line-by-line explanation**
- `a = [1, 2, 3]` → creates one list in memory and names it `a`.
- `b = a` → this does **not** create a new list. It makes `b` point to the exact **same list object** as `a`. There is only one list, with two names.
- `b[0] = 99` → since `a` and `b` refer to the same list, changing it through `b` also changes it when viewed through `a`.
- Both `print(a)` and `print(b)` show `[99, 2, 3]` — proving they are the same object.

### Question 2
Create a shallow copy using `a.copy()` and change a top-level item.

````python
a = [1, 2, 3]
b = a.copy()

b[0] = 99

print(a)
print(b)
````

**Line-by-line explanation**
- `b = a.copy()` → `.copy()` creates a **new, independent list** with the same items — this time, `a` and `b` are two separate objects.
- `b[0] = 99` → this change only affects `b`.
- `print(a)` → shows the original list, untouched: `[1, 2, 3]`.
- `print(b)` → shows the modified copy: `[99, 2, 3]`. Because the items here are simple numbers (not nested lists), a shallow copy behaves just like a full copy.

### Question 3
Create a shallow copy using slicing.

````python
a = [1, 2, 3]
b = a[:]

b[0] = 99

print(a)
print(b)
````

**Line-by-line explanation**
- `b = a[:]` → slicing with no `start` or `stop` (`[:]`) returns a copy of the **entire list**. This is another common way to create a shallow copy, alongside `.copy()`.
- The rest behaves exactly like Question 2: `b[0] = 99` only affects `b`, leaving `a` as `[1, 2, 3]` and `b` as `[99, 2, 3]`.

### Question 4
Create a nested list and demonstrate the shared inner list behavior of a shallow copy.

````python
a = [[1, 2], 3]
b = a.copy()

b[0][0] = 99

print(a)
print(b)
````

**Line-by-line explanation**
- `a = [[1, 2], 3]` → the first item of `a` is itself a **list** (`[1, 2]`) — this is a nested list.
- `b = a.copy()` → `.copy()` creates a new **outer** list, but it does **not** create new copies of the objects inside it. The inner list `[1, 2]` is still shared between `a` and `b`.
- `b[0][0] = 99` → this reaches into the shared inner list and changes it.
- Because that inner list is shared, `print(a)` shows `[[99, 2], 3]` too — even though we only modified `b`! This is exactly why a shallow copy can be misleading with nested data.

### Question 5
Use `copy.deepcopy()` with a nested list and demonstrate that changing the copy does not change the original.

````python
import copy

a = [[1, 2], 3]
b = copy.deepcopy(a)

b[0][0] = 99

print(a)
print(b)
````

**Line-by-line explanation**
- `import copy` → loads Python's built-in `copy` module, which provides the `deepcopy()` function.
- `b = copy.deepcopy(a)` → unlike `.copy()`, `deepcopy()` recursively copies **every nested object** too — so the inner list `[1, 2]` is fully duplicated, not shared.
- `b[0][0] = 99` → this only affects `b`'s own independent inner list.
- `print(a)` → stays exactly as it was: `[[1, 2], 3]`.
- `print(b)` → shows the modified copy: `[[99, 2], 3]`. This proves `a` and `b` are now completely independent, even at the nested level.

## Mini Challenge — To-Do List Manager

**Requirements recap:** a list of pending tasks, a list of completed tasks, a way to move a task from pending to completed, a way to add a new task, a way to remove a task, and clear output showing pending and completed tasks.

````python
pending_tasks = ["Praying Fajr", "Running", "Breakfast", "Studying"]
completed_tasks = []

# 1. Add a new task
pending_tasks.append("Grocery shopping")

# 2. Move a task from pending to completed
task_to_complete = pending_tasks.pop(0)
completed_tasks.append(task_to_complete)

# 3. Remove a task that is no longer needed
pending_tasks.remove("Studying")

# 4. Clear output
print("Pending tasks: ", pending_tasks)
print("Completed tasks:", completed_tasks)
````

**Line-by-line explanation**
- `pending_tasks = [...]` → creates the starting list of tasks that still need to be done.
- `completed_tasks = []` → creates an empty list that will hold tasks once they're finished.
- `pending_tasks.append("Grocery shopping")` → **adding a new task**: `.append()` adds it to the end of the pending list.
- `task_to_complete = pending_tasks.pop(0)` → **moving a task**: `.pop(0)` removes the first pending task (`"Praying Fajr"`) and *returns* it, which we store in `task_to_complete`.
- `completed_tasks.append(task_to_complete)` → that returned task is now added to the completed list — this two-step "pop then append" pattern is exactly how you move an item between two lists.
- `pending_tasks.remove("Studying")` → **removing a task**: `.remove()` deletes `"Studying"` from the pending list by value, e.g. if a task is cancelled rather than completed.
- The two `print()` statements give **clear output**, showing the current state of both lists after every operation.

**Sample output:**
```text
Pending tasks:  ['Running', 'Breakfast', 'Grocery shopping']
Completed tasks: ['Praying Fajr']
```
