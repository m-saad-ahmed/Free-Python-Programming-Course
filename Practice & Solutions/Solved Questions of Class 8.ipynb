# Encapsulation

## Question 76

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print("Balance:", account.get_balance())
```

### Line-by-line explanation

1. `class BankAccount:` — defines the account class.
2. `def __init__(self, balance):` — defines the constructor.
3. `self.__balance = balance` — stores balance in a double-underscore attribute, which Python name-mangles.
4. `def deposit(self, amount):` — defines controlled access for adding money.
5. `if amount > 0:` — validates that the deposit is positive.
6. `self.__balance += amount` — updates the protected balance internally.
7. `def get_balance(self):` — defines a method for reading the balance.
8. `return self.__balance` — returns the internal balance.
9. `account = BankAccount(1000)` — creates an account with an initial balance.
10. `account.deposit(500)` — changes the balance through the controlled method.
11. `print("Balance:", account.get_balance())` — reads the balance through the public method.

# Abstraction with Abstract Base Classes

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

square = Square(5)
print("Area:", square.area())
```

### Line-by-line explanation

2. `class Shape(ABC):` — defines an abstract base class.
3. `@abstractmethod` — marks `area()` as required for child classes.
4. `def area(self):` — declares the required method.
5. `pass` — leaves the abstract method without a concrete implementation.
6. `class Square(Shape):` — creates a concrete child class.
7. `def __init__(self, side):` — defines the square constructor.
8. `self.side = side` — stores the side length.
9. `def area(self):` — implements the required abstract method.
10. `return self.side ** 2` — calculates the square’s area.
11. `square = Square(5)` — creates a concrete Square object.
12. `print("Area:", square.area())` — calls the implemented method and prints 25.
