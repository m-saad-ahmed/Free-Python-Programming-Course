# A Manual `crop_yield` Function

## Question 6

## Question 7

**Use `crop_yield` to calculate its predicted yield.**

```python
def crop_yield(region, weights):
    result = 0
    for x, w in zip(region, weights):
        result += x * w
    return result

weights = [0.3, 0.2, 0.5]
print("Sixth-city predicted yield:", crop_yield(sixth_city, weights))
```

### 🔍 Line-by-Line Explanation

1. `def crop_yield(region, weights):` defines a reusable function with two inputs.
2. `result = 0` starts the running total at zero.
3. `zip(region, weights)` pairs each climate value with its matching weight.
4. `result += x * w` calculates each weighted contribution and adds it to the total.
5. `return result` sends the completed prediction back to the caller.
6. `crop_yield(sixth_city, weights)` applies the function to the sixth city.

## Question 8

**Explain in your own words what `zip()` does inside the `crop_yield` function.**

```python
region = [78, 85, 60]
weights = [0.3, 0.2, 0.5]

pairs = list(zip(region, weights))
print(pairs)
```

### 🔍 Line-by-Line Explanation

1. `region` contains the climate values and `weights` contains the matching weights.
2. `zip(region, weights)` pairs elements by position: first with first, second with second, and so on.
3. `list(...)` converts the zip object into a visible list of pairs.
4. The output makes it easy to see exactly which value is multiplied by which weight.

### 🔍 Line-by-Line Explanation

1. `bad_weights` intentionally contains only two weights for three climate values.
2. `crop_yield(sixth_city, bad_weights)` still runs because `zip()` stops when the shorter input ends.
3. The third climate value is therefore ignored instead of producing an automatic error.

**Explain why a silently wrong answer can be more dangerous than a program that crashes.**

# Performance — Loops vs. NumPy

## Question 16

**Repeat the timing comparison with vectors of 10,000,000 elements instead of 1,000,000.**

```python
import time
import numpy as np

n = 10_000_000
list_a = list(range(n))
list_b = list(range(n))
arr_a = np.arange(n)
arr_b = np.arange(n)

start = time.perf_counter()
list_result = [a + b for a, b in zip(list_a, list_b)]
python_time = time.perf_counter() - start

del list_result, list_a, list_b

start = time.perf_counter()
array_result = arr_a + arr_b
numpy_time = time.perf_counter() - start

del array_result

print(f"Python list loop time: {python_time:.6f} seconds")
print(f"NumPy addition time:   {numpy_time:.6f} seconds")
print(f"Speed ratio:            {python_time / numpy_time:.2f}x")
```

### 🔍 Line-by-Line Explanation

1. `n = 10_000_000` sets the requested vector size.
2. `list_a` and `list_b` create two Python lists containing ten million integers.
3. `arr_a` and `arr_b` create equivalent NumPy arrays.
4. `time.perf_counter()` records high-resolution timing around each operation.
5. The list comprehension performs addition through Python-level iteration.
6. `arr_a + arr_b` performs the same element-wise addition using NumPy’s optimized implementation.
7. `del` removes large temporary results after timing to reduce unnecessary memory use.
8. The final lines report both times and their ratio. Exact timings depend on the computer running the notebook.

## Question 17

**Predict, before running it, whether the speed gap will grow or shrink with more data — then check.**

```python
import time
import numpy as np

n = 10_000_000
arr_a = np.arange(n)
arr_b = np.arange(n)

print("Prediction: with more elements, the NumPy advantage should generally become more noticeable because Python performs more per-element interpreter work.")

start = time.perf_counter()
python_sum = sum(a + b for a, b in zip(arr_a.tolist(), arr_b.tolist()))
python_time = time.perf_counter() - start

start = time.perf_counter()
numpy_sum = np.sum(arr_a + arr_b)
numpy_time = time.perf_counter() - start

print(f"Python-style time: {python_time:.6f} seconds")
print(f"NumPy time:        {numpy_time:.6f} seconds")
print("Same result:", python_sum == numpy_sum)
```

### 🔍 Line-by-Line Explanation

1. `n = 10_000_000` creates a large workload.
2. The first `print()` records the prediction before timing, as requested.
3. The generator expression performs the addition one pair at a time through Python iteration.
4. `np.sum(arr_a + arr_b)` lets NumPy handle the array operation and reduction.
5. The timing values test the prediction; they can vary between computers.
6. Comparing `python_sum` and `numpy_sum` verifies that both approaches calculated the same total.

## Question 34

**Add a third column that is the sum of the first two, using `.reshape()` and `np.concatenate`.**

```python
import numpy as np

data_loaded = np.genfromtxt("class9_synthetic_100.csv", delimiter=",", skip_header=1)
row_sums = data_loaded.sum(axis=1).reshape(-1, 1)
combined = np.concatenate([data_loaded, row_sums], axis=1)

print("Combined shape:", combined.shape)
print(combined[:5])
```

### 🔍 Line-by-Line Explanation

1. `data_loaded.sum(axis=1)` adds the two columns separately for every row, producing 100 values.
2. `.reshape(-1, 1)` changes those 100 values from shape `(100,)` to `(100, 1)` so they form a proper column.
3. `np.concatenate([...], axis=1)` joins the original two columns and the new column horizontally.
4. The resulting array therefore has three columns.
5. `combined[:5]` previews the first five rows.

## Question 35

**Save the final combined table to a new CSV file with an appropriate header.**

```python
import numpy as np

final_path = "class9_synthetic_100_with_sum.csv"
np.savetxt(
    final_path,
    combined,
    delimiter=",")

# Rewrite with a readable header using the same numerical data.
np.savetxt(
    final_path,
    combined,
    delimiter=",")

# Add the header without changing the numerical values.
with open(final_path, "r") as f:
    body = f.read()
with open(final_path, "w") as f:
    f.write("feature_1,feature_2,row_sum\n")
    f.write(body)

print("Saved final table:", final_path)
print("Final shape:", combined.shape)
```

### 🔍 Line-by-Line Explanation

1. `final_path` stores the name of the final CSV file.
2. `np.savetxt(...)` writes the three numerical columns using commas.
3. The file is then reopened in text mode so a clean header can be placed at the top.
4. `f.write("feature_1,feature_2,row_sum\n")` adds meaningful names for all three columns.
5. `f.write(body)` writes the numerical rows underneath the header.
6. The final prints confirm the filename and the `(100, 3)` table shape.

## Question 42

**Deliberately create a broadcasting error and read the error message carefully.**

```python
a = np.ones((4, 3))
b = np.ones(2)

try:
    result = a + b
    print(result)
except ValueError as e:
    print("Expected broadcasting error:", e)
```

### 🔍 Line-by-Line Explanation

1. `a` has shape `(4, 3)`.
2. `b` has shape `(2,)`.

## Question 43

**Fix your broadcasting error by adjusting the shape of one of the arrays.**

```python
a = np.ones((4, 3))
b = np.ones(2)

# Fix: use a vector whose length matches the number of columns.
b_fixed = np.ones(3)
result = a + b_fixed

print("Fixed shapes:", a.shape, b_fixed.shape)
print(result)
```

### 🔍 Line-by-Line Explanation

1. `a` remains a `(4, 3)` array.
2. `b_fixed = np.ones(3)` creates three values, matching the three columns of `a`.
3. `a + b_fixed` can now broadcast the vector across all four rows.
4. The printed shapes make the compatibility clear.
5. The result contains ones plus ones, so every element becomes two.

```python
shape_a = (4, 3)
shape_b = (3,)

print("Shape A:", shape_a)
print("Shape B:", shape_b)
print("Compatible: yes — the trailing dimensions are equal (3 and 3).")
```

### 🔍 Line-by-Line Explanation

1. `shape_a` represents a 4×3 array.
2. `shape_b` represents a one-dimensional array with three elements.
3. NumPy compares dimensions from right to left.

## Question 45

**Explain why broadcasting is more memory-efficient than manually copying an array to match another array's shape.**

```python
large = np.zeros((1000, 1000))
row = np.ones(1000)

broadcasted_result = large + row
manual_copy = np.broadcast_to(row, large.shape)

print("Large array shape:", large.shape)
print("Row shape:", row.shape)
print("Broadcast view shape:", manual_copy.shape)
print("Results match:", np.array_equal(broadcasted_result, manual_copy))
```

### 🔍 Line-by-Line Explanation

1. `large` is a 1000×1000 array.
2. `row` contains only 1000 values.
3. NumPy can apply `row` across all rows without requiring the programmer to build 1000 separate copies.
4. `np.broadcast_to` demonstrates the matching broadcasted shape without constructing an ordinary full copy of the data.
5. This avoids unnecessary memory use and is one reason broadcasting is both convenient and efficient.
