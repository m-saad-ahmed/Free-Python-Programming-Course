# Question 5 — Histogram

## Question

Generate 300 numerical observations and compare two different bin counts.


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
observations = rng.normal(loc=50, scale=10, size=300)

fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)

axes[0].hist(observations, bins=10)
axes[0].set_title("10 Bins")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Frequency")

axes[1].hist(observations, bins=30)
axes[1].set_title("30 Bins")
axes[1].set_xlabel("Value")

fig.suptitle("Same Data, Different Bin Counts")
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

2. `default_rng(42)` creates a reproducible random-number generator.
3. `normal(...)` generates 300 observations around a mean of 50 with a standard deviation of 10.
4. `plt.subplots(1, 2, ...)` creates two Axes side by side.
5. The first `hist(..., bins=10)` groups the same data into 10 intervals.
6. The second histogram uses 30 intervals, producing a more detailed view.
7. The titles and labels identify the two choices.
8. `sharey=True` keeps the frequency scale comparable.
9. `fig.suptitle(...)` adds one overall title.

# Question 7 — Annotation

## Question

Annotate the maximum point in a line chart.


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
values = [18, 24, 21, 31, 27, 35]

max_index = values.index(max(values))
max_value = values[max_index]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(months, values, marker="o")
ax.annotate(
    f"Maximum: {max_value}",
    xy=(max_index, max_value),
    xytext=(max_index - 1, max_value - 8),
    arrowprops={"arrowstyle": "->"}
)
ax.set_title("Trend with Maximum Point Annotated")
ax.set_xlabel("Month")
ax.set_ylabel("Value")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. `months` and `values` contain the x- and y-data.
2. `max(values)` finds the largest value.
3. `.index(...)` finds the position of that maximum value.
4. `max_value` stores the maximum for reuse.
5. `ax.plot(...)` draws the trend.
6. `ax.annotate(...)` adds text connected to a specific data point.
7. `xy` tells Matplotlib which point to annotate.
8. `xytext` specifies where the annotation text should appear.
9. `arrowprops` draws an arrow from the text toward the data point.
10. The remaining statements provide chart context and display the Figure.

# Question 11 — Correlation

## Question

Create a DataFrame with at least four numerical variables and visualize its correlation matrix.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(21)
study = rng.uniform(1, 8, 40)
sleep = rng.uniform(5, 9, 40)
attendance = rng.uniform(60, 100, 40)
score = 45 + 5 * study + 1.2 * sleep + 0.25 * attendance + rng.normal(0, 8, 40)

df = pd.DataFrame({
    "Study Hours": study,
    "Sleep Hours": sleep,
    "Attendance": attendance,
    "Score": score
})

corr = df.corr()

fig, ax = plt.subplots(figsize=(7, 6))
image = ax.imshow(corr, vmin=-1, vmax=1)
fig.colorbar(image, ax=ax, label="Correlation")
ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation=45, ha="right")
ax.set_yticklabels(corr.columns)
ax.set_title("Correlation Matrix")

for i in range(len(corr)):
    for j in range(len(corr)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")

plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. NumPy generates reproducible synthetic variables.
2. `score` is constructed from several variables plus random noise so the dataset contains measurable relationships.
3. `pd.DataFrame({...})` combines the variables into a tabular structure.
4. `df.corr()` calculates pairwise Pearson correlations between numerical columns.
5. `imshow(corr, vmin=-1, vmax=1)` visualizes the correlation matrix using the fixed correlation range from -1 to +1.
6. `fig.colorbar(...)` explains the visual scale.
7. `set_xticks` and `set_yticks` create positions for the variable names.
8. `set_xticklabels` and `set_yticklabels` display those names.
9. The nested loops write the numerical correlation value into every cell.
10. `tight_layout()` prevents labels from being clipped, and `show()` displays the result.


# Question 12 — pcolormesh()

## Question

Create a coordinate grid and visualize a continuous mathematical function.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 80)
y = np.linspace(-3, 3, 80)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

fig, ax = plt.subplots(figsize=(7, 5))
mesh = ax.pcolormesh(X, Y, Z, shading="auto")
fig.colorbar(mesh, ax=ax, label="f(x, y)")
ax.set_title("Continuous Function: sin(x) · cos(y)")
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. `linspace` creates evenly spaced x- and y-coordinates.
2. `meshgrid` turns the one-dimensional coordinates into a two-dimensional coordinate grid.
3. `Z = ...` evaluates the mathematical function at every grid location.
4. `pcolormesh(X, Y, Z, shading="auto")` draws colored cells representing the function values.
5. The returned mesh object is saved so it can be connected to a colorbar.
6. The colorbar represents the numerical value of the function.
7. The title and axis labels identify the mathematical variables.
8. `tight_layout()` and `show()` finish the visualization.


# Question 13 — fill_between()

## Question

Create a trend with a lower and upper uncertainty bound.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 13)
trend = 40 + 2.2 * x + 4 * np.sin(x / 2)
uncertainty = 2.5 + 0.15 * x
lower = trend - uncertainty
upper = trend + uncertainty

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(x, trend, marker="o", label="Trend")
ax.fill_between(x, lower, upper, alpha=0.25, label="Uncertainty band")
ax.set_title("Trend with Uncertainty Band")
ax.set_xlabel("Month")
ax.set_ylabel("Value")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. `x` represents twelve months.
2. `trend` creates a changing central estimate.
3. `uncertainty` defines how wide the uncertainty is at each month.
4. `lower` and `upper` calculate the two boundaries.
5. `ax.plot(...)` draws the central trend.
6. `fill_between(x, lower, upper)` shades the region between the two boundaries.
7. `label=...` gives both plotted elements names for the legend.
8. The title and labels explain the visualization.
9. `legend()` identifies the line and band.
10. The grid, layout adjustment, and `show()` complete the chart.

# Question 16 — twinx()

## Question

Create one chart containing revenue and growth percentage on separate y-axes.


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [120, 135, 128, 150, 172, 185]
growth = [4, 7, -2, 10, 8, 6]

fig, ax1 = plt.subplots(figsize=(9, 4))
ax2 = ax1.twinx()

ax1.plot(months, revenue, marker="o", label="Revenue")
ax2.plot(months, growth, marker="s", label="Growth %")

ax1.set_xlabel("Month")
ax1.set_ylabel("Revenue")
ax2.set_ylabel("Growth (%)")
ax1.set_title("Revenue and Growth Percentage")

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper left")

ax1.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. `revenue` and `growth` use the same x-axis categories but have different units and scales.
2. `plt.subplots()` creates the main Axes.
3. `ax1.twinx()` creates a second y-axis sharing the same x-axis.
4. `ax1.plot(...)` draws revenue against the left y-axis.
5. `ax2.plot(...)` draws growth against the right y-axis.
6. Each y-axis gets its own label because the units differ.
7. `get_legend_handles_labels()` collects legend entries from both Axes.
8. Combining those entries creates one legend describing both lines.

# Question 17 — twiny()

## Question

Create two related x-axis scales sharing one y-axis.


```python
import numpy as np
import matplotlib.pyplot as plt

temperature_c = np.array([0, 10, 20, 30, 40])
time_hours = np.array([0, 2, 4, 6, 8])

fig, ax1 = plt.subplots(figsize=(8, 4))
ax2 = ax1.twiny()

ax1.plot(temperature_c, [5, 12, 20, 29, 39], marker="o")
ax1.set_xlabel("Temperature (°C)")
ax1.set_ylabel("Measured Response")
ax1.set_title("Two Related X-Axis Scales")

ax2.set_xlim(ax1.get_xlim())
ax2.set_xticks(temperature_c)
ax2.set_xticklabels([f"{t}°C → {h}h" for t, h in zip(temperature_c, time_hours)])
ax2.set_xlabel("Equivalent Time Scale")

ax1.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. `temperature_c` provides the primary x-axis scale.
2. `time_hours` provides a related second scale.
3. The response values are plotted against temperature on `ax1`.
4. `ax1.twiny()` creates a second x-axis while keeping the same vertical axis.
5. `ax2.set_xlim(ax1.get_xlim())` aligns the second x-axis with the first.
6. The second axis uses the same x positions but displays related labels.
7. `set_xticklabels(...)` formats the second scale as paired temperature-to-time labels.
8. Both axes share the same y values, while their x labels communicate two related interpretations.

# Question 19 — GridSpec

## Question

Design a dashboard containing one large chart and three smaller charts.


```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
import numpy as np

x = np.arange(1, 13)
sales = [20, 24, 23, 29, 31, 30, 35, 38, 37, 42, 45, 48]
categories = ["A", "B", "C", "D"]
category_values = [35, 28, 22, 15]
study = [1, 2, 3, 4, 5, 6]
scores = [52, 58, 64, 72, 81, 88]

fig = plt.figure(figsize=(12, 8))
gs = GridSpec(2, 2, figure=fig)

ax_big = fig.add_subplot(gs[0, :])
ax_bar = fig.add_subplot(gs[1, 0])
ax_scatter = fig.add_subplot(gs[1, 1])

ax_big.plot(x, sales, marker="o")
ax_big.set_title("Large Trend Chart")
ax_big.set_xlabel("Month")
ax_big.set_ylabel("Sales")
ax_big.grid(True, alpha=0.3)

ax_bar.bar(categories, category_values)
ax_bar.set_title("Category Comparison")
ax_bar.set_ylabel("Value")

ax_scatter.scatter(study, scores)
ax_scatter.set_title("Study vs Score")
ax_scatter.set_xlabel("Study Hours")
ax_scatter.set_ylabel("Score")

fig.suptitle("Matplotlib GridSpec Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. `GridSpec` is imported to control the layout precisely.

# Question 20 — 3D

## Question

Create a mathematical 3D surface using `mplot3d`.


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-3, 3, 60)
y = np.linspace(-3, 3, 60)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z)
ax.set_title("3D Mathematical Surface")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.tight_layout()
plt.show()
```

## Line-by-Line Explanation

1. `linspace` creates evenly spaced x- and y-coordinates.
2. `meshgrid` creates the two-dimensional coordinate system needed for the surface.
3. `Z` evaluates a mathematical function at every coordinate.
4. `plt.figure()` creates the 3D Figure.
5. `fig.add_subplot(..., projection="3d")` creates a three-dimensional Axes.
6. `plot_surface(X, Y, Z)` renders the mathematical surface.
7. The three axis labels identify X, Y, and Z.

# Question 26 — Complete Visualization Report

## Question

Choose a dataset of your own.

Your notebook must include:
- a clear analytical question
- data inspection
- at least three appropriate charts
- one advanced visualization technique
- correct labels and units
- at least one annotation
- a professional layout
- saved output
- a written interpretation
- a short limitations section

The objective is to communicate useful information correctly and clearly.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a small dataset of monthly business performance.
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug"],
    "Sales": [120, 128, 125, 142, 150, 158, 171, 176],
    "Customers": [48, 51, 49, 56, 60, 63, 69, 72],
    "Satisfaction": [78, 80, 79, 82, 84, 85, 87, 88]
})

# 2. Analytical question.
print("Analytical question:")
print("How did sales change over time, and is higher customer volume associated with higher sales?")

# 3. Data inspection.
print("\nData preview:")
print(df.head())
print("\nData types:")
print(df.dtypes)
print("\nSummary statistics:")
print(df[["Sales", "Customers", "Satisfaction"]].describe())

# 4. Prepare numeric x positions.
x = np.arange(len(df))

# 5. Create a professional multi-chart report.
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Chart 1: trend.
axes[0, 0].plot(x, df["Sales"], marker="o")
axes[0, 0].set_title("Monthly Sales Trend")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Sales (units)")
axes[0, 0].set_xticks(x)
axes[0, 0].set_xticklabels(df["Month"])
axes[0, 0].grid(True, alpha=0.3)

# Annotation.
max_index = df["Sales"].idxmax()
axes[0, 0].annotate(
    f"Peak: {df.loc[max_index, 'Sales']} units",
    xy=(max_index, df.loc[max_index, "Sales"]),
    xytext=(max_index - 2, df.loc[max_index, "Sales"] - 25),
    arrowprops={"arrowstyle": "->"}
)

# Chart 2: category/metric comparison.
axes[0, 1].bar(df["Month"], df["Customers"])
axes[0, 1].set_title("Customers by Month")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Customers")
axes[0, 1].grid(axis="y", alpha=0.3)

# Chart 3: relationship.
axes[1, 0].scatter(df["Customers"], df["Sales"])
axes[1, 0].set_title("Customers vs Sales")
axes[1, 0].set_xlabel("Customers")
axes[1, 0].set_ylabel("Sales (units)")
axes[1, 0].grid(True, alpha=0.3)

# Advanced visualization: uncertainty-style band around sales.
sales = df["Sales"].to_numpy()
band = np.array([5, 5, 6, 6, 7, 7, 8, 8])
axes[1, 1].plot(x, sales, marker="o", label="Sales")
axes[1, 1].fill_between(x, sales - band, sales + band, alpha=0.25, label="Uncertainty-style band")
axes[1, 1].set_title("Sales with Uncertainty-Style Band")
axes[1, 1].set_xlabel("Month")
axes[1, 1].set_ylabel("Sales (units)")
axes[1, 1].set_xticks(x)
axes[1, 1].set_xticklabels(df["Month"])
axes[1, 1].legend()
axes[1, 1].grid(True, alpha=0.3)

fig.suptitle("Monthly Business Performance Report", fontsize=16)
plt.tight_layout()

# 6. Save the final output.
output_file = "class12_complete_visualization_report.png"
fig.savefig(output_file, dpi=150, bbox_inches="tight")
plt.show()

# 7. Written interpretation.
print("\nInterpretation:")
print("Sales show an overall upward trend from January to August, with a small dip in March.")
print("Customer counts also increase overall, and the scatter plot shows a strong positive association with sales in this small dataset.")
print("The uncertainty-style band communicates a range around each sales value, but it is illustrative rather than a statistically estimated confidence interval.")

# 8. Limitations.
print("\nLimitations:")
print("This is a small synthetic dataset covering only eight months.")
print("The observed association between customers and sales does not establish causation.")
print("The uncertainty band is illustrative because no sampling model or measurement-error calculation was provided.")
```
