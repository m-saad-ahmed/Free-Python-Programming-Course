## Question 1

**Explain, in your own words, why Pandas is useful for tabular data and what role a DataFrame plays. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Pandas is used for labeled tabular data; a DataFrame stores rows and columns with aligned labels. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"name":["Ali","Sara","Hamza"],"score":[85,92,78]})
print(df)
print("shape:", df.shape)
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain the difference between a Series' values, index, name, and dtype. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** A Series is a one-dimensional labeled array with values, an index, a name, and a dtype. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s = pd.Series([85,92,78], index=["Ali","Sara","Hamza"], name="score")
print(s)
print("Sara:", s.loc["Sara"])
print("first:", s.iloc[0])
print("dtype:", s.dtype)
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain why a DataFrame can be understood as aligned Series objects. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** A DataFrame is a two-dimensional table whose columns behave like aligned Series. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"name":["Ali","Sara","Hamza"],"score":[85,92,78]})
print(type(df["score"]).__name__)
print(type(df[["name","score"]]).__name__)
print(df.iloc[0])
print(df.iloc[[0,2]])
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.
2. `df = pd.DataFrame({"name":["Ali","Sara","Hamza"],"score":[85,92,78]})` — `df = pd.DataFrame({"name":["Ali","Sara","Hamza"],"score":[85,92,78]})` `df` mein operation ka result store karta hai.
3. `print(type(df["score"]).__name__)` — `print(type(df["score"]).__name__)` relevant result/validation ko notebook output mein display karta hai.
4. `print(type(df[["name","score"]]).__name__)` — `print(type(df[["name","score"]]).__name__)` relevant result/validation ko notebook output mein display karta hai.
5. `print(df.iloc[0])` — `print(df.iloc[0])` relevant result/validation ko notebook output mein display karta hai.
6. `print(df.iloc[[0,2]])` — `print(df.iloc[[0,2]])` relevant result/validation ko notebook output mein display karta hai.

# Module 4 — Module 4 — Index and Labels

---

## Question 1

**Explain the difference between labels and positions in Pandas. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Pandas distinguishes labels from integer positions; loc is label-based and iloc is position-based. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"score":[85,92,78]}, index=["A101","A102","A103"])
print(df.loc["A102"])
print(df.iloc[1])
print(df.reset_index())
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain why loading a file is only the first step of a data workflow. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Reading/writing data connects Pandas to files; a reproducible workflow keeps file paths and transformations explicit. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
from pathlib import Path
path=Path("pandas_demo.csv")
df=pd.DataFrame({"name":["Ali","Sara"],"score":[85,92]})
df.to_csv(path,index=False)
loaded=pd.read_csv(path)
print(loaded)
print("same shape:", loaded.shape==df.shape)
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 8 — Module 8 — Boolean Filtering

---

## Question 1

**Explain how a Boolean mask controls which rows survive a filter. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Boolean masks are True/False arrays aligned to rows; only True rows are kept. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df=pd.DataFrame({"name":["Ali","Sara","Hamza","Zoya"],"score":[85,92,78,88],"city":["K","L","K","I"]})
mask=(df["score"]>=85)&(df["city"].isin(["K","L"]))
print(df[mask])
print("kept:",mask.sum())
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 14 — Module 14 — Missing Data

---

## Question 1

**Explain why a missing value should not automatically be treated as zero. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Missing values mean information is absent; replacing them with zero can create false measurements. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df=pd.DataFrame({"name":["Ali","Sara","Hamza","Zoya"],"score":[85,None,78,None]})
print("before:\n",df.isna().sum())
df["score"]=df["score"].fillna(df["score"].median())
print("after:\n",df)
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.

3. `print("before:\n",df.isna().sum())` — `print("before:\n",df.isna().sum())` relevant result/validation ko notebook output mein display karta hai.
4. `df["score"]=df["score"].fillna(df["score"].median())` — `df["score"]=df["score"].fillna(df["score"].median())` missing values ko specified replacement se fill karta hai.

# Module 15 — Module 15 — Duplicates

---

## Question 1

**Explain why defining the business key matters before removing duplicates. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Duplicate handling should follow a defined business key rather than blindly deleting identical-looking rows. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df=pd.DataFrame({"order_id":[1,1,2,3],"item":["Book","Book","Pen","Bag"]})
print("duplicates:",df.duplicated().sum())
print("by key:",df.duplicated(subset=["order_id","item"]).sum())
print(df.drop_duplicates(subset=["order_id","item"]))
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain the difference between `unique()`, `nunique()`, and `value_counts()`. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** unique lists distinct values, nunique counts them, and value_counts gives frequencies. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s=pd.Series(["Gold","Silver","Gold","Bronze","Gold","Silver"])
print("unique:",s.unique().tolist())
print("nunique:",s.nunique())
print("counts:\n",s.value_counts())
print("normalized:\n",s.value_counts(normalize=True))
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain the SPLIT → APPLY → COMBINE pattern in your own words. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** GroupBy follows SPLIT → APPLY → COMBINE: partition rows, calculate within groups, then return a summary. The practical rule is to understand the input structure and expected output before executing the operation.

# Module 21 — Module 21 — Advanced GroupBy

---

## Question 1

**Explain the difference between `agg()` and `transform()` in terms of output shape. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** agg usually reduces each group to summary rows, while transform returns group-derived values aligned to original rows. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df=pd.DataFrame({"city":["K","K","L","L"],"sales":[100,150,80,120]})
print(df.groupby("city")["sales"].agg(["min","mean","max"]))
df["city_mean"]=df.groupby("city")["sales"].transform("mean")
print(df)
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.
2. `df=pd.DataFrame({"city":["K","K","L","L"],"sales":[100,150,80,120]})` — `df=pd.DataFrame({"city":["K","K","L","L"],"sales":[100,150,80,120]})` `df` mein operation ka result store karta hai.
3. `print(df.groupby("city")["sales"].agg(["min","mean","max"]))` — `print(df.groupby("city")["sales"].agg(["min","mean","max"]))` relevant result/validation ko notebook output mein display karta hai.
4. `df["city_mean"]=df.groupby("city")["sales"].transform("mean")` — `df["city_mean"]=df.groupby("city")["sales"].transform("mean")` rows ko grouping key ke mutabiq groups mein divide karke grouped operation apply karta hai.
5. `print(df)` — `print(df)` relevant result/validation ko notebook output mein display karta hai.
6. `print("✓ Example executed successfully; inspect the output against the question.")` — `print("✓ Example executed successfully; inspect the output against the question.")` relevant result/validation ko notebook output mein display karta hai.

## Question 1

**Explain the difference between row-wise and column-wise concatenation. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** concat combines objects along an axis; row-wise stacking appends observations, while column-wise concatenation aligns indexes. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
jan=pd.DataFrame({"month":["Jan","Jan"],"sales":[100,120]})
feb=pd.DataFrame({"month":["Feb","Feb"],"sales":[130,110]})
print(pd.concat([jan,feb],ignore_index=True))
a=pd.DataFrame({"sales":[100,120]},index=["A","B"])
b=pd.DataFrame({"cost":[60,70]},index=["B","C"])
print(pd.concat([a,b],axis=1))
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 25 — Module 25 — Pivot and Reshaping

---

## Question 1

**Explain what reshaping changes and what it should preserve. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Reshaping changes the arrangement of observations while preserving their information; melt makes wide data long and pivot makes long data wide. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
wide=pd.DataFrame({"product":["A","B"],"Jan":[100,120],"Feb":[110,130]})
long=wide.melt(id_vars="product",var_name="month",value_name="sales")
print(long)
print(long.pivot(index="product",columns="month",values="sales"))
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

3. `long=wide.melt(id_vars="product",var_name="month",value_name="sales")` — `long=wide.melt(id_vars="product",var_name="month",value_name="sales")` wide columns ko long rows mein reshape karta hai.
4. `print(long)` — `print(long)` relevant result/validation ko notebook output mein display karta hai.
5. `print(long.pivot(index="product",columns="month",values="sales"))` — `print(long.pivot(index="product",columns="month",values="sales"))` relevant result/validation ko notebook output mein display karta hai.

# Module 27 — Module 27 — Indexing and MultiIndex

---

## Question 1

**Explain why indexes matter for alignment as well as selection. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Indexes support alignment and selection; MultiIndex stores hierarchical labels across levels. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
idx=pd.MultiIndex.from_product([["K","L"],["Jan","Feb"]],names=["city","month"])
df=pd.DataFrame({"sales":[100,120,80,90]},index=idx)
print(df)
print(df.xs("K",level="city"))
print(df.reset_index())
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.
2. `idx=pd.MultiIndex.from_product([["K","L"],["Jan","Feb"]],names=["city","month"])` — `idx=pd.MultiIndex.from_product([["K","L"],["Jan","Feb"]],names=["city","month"])` `idx` mein operation ka result store karta hai.
3. `df=pd.DataFrame({"sales":[100,120,80,90]},index=idx)` — `df=pd.DataFrame({"sales":[100,120,80,90]},index=idx)` `df` mein operation ka result store karta hai.
4. `print(df)` — `print(df)` relevant result/validation ko notebook output mein display karta hai.
5. `print(df.xs("K",level="city"))` — `print(df.xs("K",level="city"))` relevant result/validation ko notebook output mein display karta hai.
6. `print(df.reset_index())` — `print(df.reset_index())` relevant result/validation ko notebook output mein display karta hai.
7. `print("✓ Example executed successfully; inspect the output against the question.")` — `print("✓ Example executed successfully; inspect the output against the question.")` relevant result/validation ko notebook output mein display karta hai.

# Module 30 — Module 30 — Resampling and Rolling Windows

---

## Question 1

**Explain the difference between resampling and rolling windows. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Resampling groups observations into calendar bins; rolling windows calculate moving statistics over neighboring observations. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
idx=pd.date_range("2026-01-01",periods=6,freq="D")
df=pd.DataFrame({"sales":[10,20,15,30,25,40]},index=idx)
print(df.resample("3D").sum())
print(df["sales"].rolling(3).mean())
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 32 — Module 32 — Outlier Handling

---

## Question 1

**Explain why an outlier is not automatically an error. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** An outlier is an unusual observation, not automatically an error; investigate its provenance before removing it. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s=pd.Series([10,11,12,13,100])
q1,q3=s.quantile(.25),s.quantile(.75)
iqr=q3-q1
mask=(s<q1-1.5*iqr)|(s>q3+1.5*iqr)
print("bounds:",q1-1.5*iqr,q3+1.5*iqr)
print("candidates:",s[mask].tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.

An outlier is an unusual observation, not automatically an error; investigate its provenance before removing it. The worked code is small but follows the same pattern used on larger datasets.

```python
import pandas as pd
s=pd.Series([10,11,12,13,100])
q1,q3=s.quantile(.25),s.quantile(.75)
iqr=q3-q1
mask=(s<q1-1.5*iqr)|(s>q3+1.5*iqr)
print("bounds:",q1-1.5*iqr,q3+1.5*iqr)
print("candidates:",s[mask].tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 35 — Module 35 — Copy vs View and Safe Assignment

---

## Question 1

**Explain why explicit copying and assignment make transformations safer. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Explicit copies and .loc assignments make it clear which object is being modified and avoid ambiguous chained assignment. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df=pd.DataFrame({"score":[40,60,80],"status":["Fail","Pass","Pass"]})
work=df[df["score"]>=50].copy()
work.loc[work["score"]>=70,"status"]="Excellent"
print("original:\n",df)
print("work:\n",work)
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.

3. `work=df[df["score"]>=50].copy()` — `work=df[df["score"]>=50].copy()` independent copy banata hai taa-ke original object ko accidentally modify na kiya jaye.
4. `work.loc[work["score"]>=70,"status"]="Excellent"` — `work.loc[work["score"]>=70,"status"]="Excellent"` label-based selection/assignment karta hai; `.loc` row/column labels ko use karta hai.
5. `print("original:\n",df)` — `print("original:\n",df)` relevant result/validation ko notebook output mein display karta hai.
6. `print("work:\n",work)` — `print("work:\n",work)` relevant result/validation ko notebook output mein display karta hai.

# Module 38 — Module 38 — Large Datasets and Performance

---

## Question 1

**Explain why loading less data can improve both memory use and workflow speed. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Performance improves when unnecessary columns/rows are avoided and vectorized operations or chunks are used appropriately. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
import numpy as np
df=pd.DataFrame({"id":range(1000),"sales":np.arange(1000,dtype=float),"unused":np.arange(1000)})
needed=df[["id","sales"]].copy()
needed["sales_plus"]=needed["sales"]+1
print("full bytes:",df.memory_usage(deep=True).sum())
print("needed bytes:",needed.memory_usage(deep=True).sum())
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.
2. `import numpy as np` — NumPy ko `np` short name ke saath import karta hai, mainly numeric demo data ke liye.
3. `df=pd.DataFrame({"id":range(1000),"sales":np.arange(1000,dtype=float),"unused":np.arange(1000)})` — `df=pd.DataFrame({"id":range(1000),"sales":np.arange(1000,dtype=float),"unused":np.arange(1000)})` `df` mein operation ka result store karta hai.

## Question 1

**Explain what makes a Pandas workflow reproducible and reviewable. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** A professional workflow makes assumptions, inputs, transformations, outputs, and validation steps explicit and reproducible. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
raw=pd.DataFrame({"name":["Ali","Sara","Hamza"],"score":[80,None,95]})
print("inspection:\n",raw.info())
clean=raw.copy()
clean["score"]=clean["score"].fillna(clean["score"].median())
result=clean.query("score>=80").sort_values("score",ascending=False)
assert result["score"].notna().all()
print(result)
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain the main mental model of `Binning with cut() and qcut()` in your own words.**

### ✅ Answer

**Answer:** cut creates fixed-width or explicitly bounded bins; qcut creates quantile-based bins. Labels and q are important controls. The key is to understand what happens to the labels, rows, columns, and values before writing code.

```python
import pandas as pd
df = pd.DataFrame({"name":["Ali","Sara","Hamza","Zoya","Bilal"],"age":[18,22,27,31,45]})
df["age_bin"] = pd.cut(df["age"], bins=[0,20,30,40,60], labels=["Teen","20s","30s","40+"])
df["age_quantile"] = pd.qcut(df["age"], q=3, labels=["Low","Mid","High"])
print(df)
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 42 — Module 42 — crosstab()

---

## Question 1

**Explain the main mental model of `crosstab()` in your own words.**

### ✅ Answer

**Answer:** crosstab summarizes combinations of categorical variables; normalize changes counts into proportions. The key is to understand what happens to the labels, rows, columns, and values before writing code.

```python
import pandas as pd
df = pd.DataFrame({"sex":["F","M","F","M","F","M"],"survived":[1,0,1,1,0,0]})
table = pd.crosstab(df["sex"], df["survived"])
rates = pd.crosstab(df["sex"], df["survived"], normalize="index")
print("counts:\n", table)
print("row proportions:\n", rates)
print("✓ Example executed successfully; inspect the output against the question.")
```

3. `table = pd.crosstab(df["sex"], df["survived"])` — `table = pd.crosstab(df["sex"], df["survived"])` `table` mein operation ka result store karta hai.
4. `rates = pd.crosstab(df["sex"], df["survived"], normalize="index")` — `rates = pd.crosstab(df["sex"], df["survived"], normalize="index")` `rates` mein operation ka result store karta hai.

**Answer:** explode turns list-like values in a cell into multiple rows; ignore_index can reset the resulting index. The key is to understand what happens to the labels, rows, columns, and values before writing code.

# Module 44 — Module 44 — pipe()

---

## Question 1

**Explain the main mental model of `pipe()` in your own words.**

### ✅ Answer

**Answer:** pipe passes a DataFrame through a function so custom transformations fit naturally into a readable pipeline. The key is to understand what happens to the labels, rows, columns, and values before writing code.

```python
import pandas as pd
def clean_sales(frame):
    return frame.assign(sales=frame["sales"].fillna(0)).query("sales > 0")
df = pd.DataFrame({"sales":[100,None,150],"city":["K","L","K"]})
result = df.pipe(clean_sales)
print(result)
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 45 — Module 45 — stack() and unstack()

## Question 1

**Explain the main mental model of `stack() and unstack()` in your own words.**

### ✅ Answer

**Answer:** stack moves column levels into the row index; unstack moves index levels into columns. The key is to understand what happens to the labels, rows, columns, and values before writing code.

```python
import pandas as pd
idx = pd.MultiIndex.from_product([["K","L"],["Jan","Feb"]], names=["city","month"])
df = pd.DataFrame({"sales":[100,120,80,90]}, index=idx)
stacked = df.unstack("month")
print("unstacked:\n", stacked)
print("restored:\n", stacked.stack())
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.

3. `df = pd.DataFrame({"sales":[100,120,80,90]}, index=idx)` — `df = pd.DataFrame({"sales":[100,120,80,90]}, index=idx)` `df` mein operation ka result store karta hai.
4. `stacked = df.unstack("month")` — `stacked = df.unstack("month")` row/column arrangement ko stack/unstack karke reshape karta hai.
5. `print("unstacked:\n", stacked)` — `print("unstacked:\n", stacked)` relevant result/validation ko notebook output mein display karta hai.
6. `print("restored:\n", stacked.stack())` — `print("restored:\n", stacked.stack())` relevant result/validation ko notebook output mein display karta hai.
7. `print("✓ Example executed successfully; inspect the output against the question.")` — `print("✓ Example executed successfully; inspect the output against the question.")` relevant result/validation ko notebook output mein display karta hai.

## Question 1

### ✅ Answer

**Answer:** nlargest and nsmallest efficiently select extreme rows; keep controls tie handling. The key is to understand what happens to the labels, rows, columns, and values before writing code.

# Module 49 — Module 49 — Excel sheet_name=

---

## Question 1

**Explain the main mental model of `Excel sheet_name=` in your own words.**

### ✅ Answer

**Answer:** read_excel uses sheet_name to choose which worksheet(s) to read from an Excel workbook. The key is to understand what happens to the labels, rows, columns, and values before writing code.

```python
import pandas as pd
from pathlib import Path
path = Path("class11_demo.xlsx")
with pd.ExcelWriter(path) as writer:
    pd.DataFrame({"name":["Ali","Sara"],"score":[80,90]}).to_excel(writer, sheet_name="Scores", index=False)
    pd.DataFrame({"city":["K","L"],"sales":[100,120]}).to_excel(writer, sheet_name="Sales", index=False)
print("sheet names:", pd.ExcelFile(path).sheet_names)
print("Scores:\n", pd.read_excel(path, sheet_name="Scores"))
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain the main mental model of `Nullable Int64 and pd.NA` in your own words.**

### ✅ Answer

**Answer:** Nullable Int64 allows integer columns to contain pd.NA without converting them to floating-point dtype. The key is to understand what happens to the labels, rows, columns, and values before writing code.

```python
import pandas as pd
s = pd.Series([1, 2, pd.NA, 4], dtype="Int64")
print(s)
print("dtype:", s.dtype)
print("is missing:", s.isna().tolist())
print("plus 1:", (s + 1).tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 52 — Module 52 — MultiIndex xs(), droplevel(), swaplevel()

---

## Question 1

**Explain the main mental model of `MultiIndex xs(), droplevel(), swaplevel()` in your own words.**

### ✅ Answer

**Answer:** xs selects across MultiIndex levels; droplevel removes a level; swaplevel changes level order. The key is to understand what happens to the labels, rows, columns, and values before writing code.

```python
import pandas as pd
idx = pd.MultiIndex.from_product([["K","L"],["Jan","Feb"]], names=["city","month"])
df = pd.DataFrame({"sales":[100,120,80,90]}, index=idx)
print("xs K:\n", df.xs("K", level="city"))
print("droplevel:\n", df.droplevel("month"))
print("swapped levels:\n", df.swaplevel("city","month"))
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

# Module 54 — Module 54 — shift()

---

## Question 1

**Explain `shift()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** shift moves values relative to their index, making lag/lead comparisons possible. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s = pd.Series([100,120,150,130], index=pd.date_range("2026-01-01", periods=4), name="sales")
print("previous:", s.shift(1).tolist())
df = pd.DataFrame({"sales":s})
df["previous_sales"] = df["sales"].shift(1)
print(df)
print("✓ Example executed successfully; inspect the output against the question.")
```

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.

# Module 58 — Module 58 — rank()

---

## Question 1

**Explain `rank()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** rank assigns relative positions to values and supports tie-handling methods. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s = pd.Series([90,70,90,80], index=["A","B","C","D"], name="score")
print("average ranks:\n", s.rank())
print("dense ranks:\n", s.rank(method="dense", ascending=False))
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

# Module 59 — Module 59 — where() / mask()

---

## Question 1

**Explain `where() / mask()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** where keeps values where a condition is true and replaces false positions; mask does the complementary condition. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s = pd.Series([40,60,80,100])
print("where >=60:", s.where(s >= 60, 0).tolist())
print("mask >=80:", s.mask(s >= 80, 0).tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

# Module 63 — Module 63 — select_dtypes() / convert_dtypes()

---

## Question 1

**Explain `select_dtypes() / convert_dtypes()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** select_dtypes chooses columns by dtype; convert_dtypes improves dtype choices using Pandas nullable types. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"name":["Ali","Sara"],"score":[80,90],"passed":[True,True]})
print("numeric:\n", df.select_dtypes(include="number"))
converted = df.convert_dtypes()
print("converted dtypes:\n", converted.dtypes)
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 64 — Module 64 — corr() / cov()

---

## Question 1

**Explain `corr() / cov()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** corr measures standardized linear association; cov measures covariance in the original units. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"hours":[2,4,6,8],"score":[50,60,80,90]})
print("correlation:\n", df.corr())
print("covariance:\n", df.cov())
print("✓ Example executed successfully; inspect the output against the question.")
```

## Question 1

**Explain `Timedelta` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** Timedelta represents a duration and supports arithmetic between timestamps and time intervals. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
start = pd.Timestamp("2026-01-01 09:00")
end = pd.Timestamp("2026-01-02 14:30")
duration = end - start
print("duration:", duration)
print("hours:", duration.total_seconds() / 3600)
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 67 — Module 67 — interpolate()

---

## Question 1

**Explain `interpolate()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** interpolate estimates missing numeric values from surrounding observations according to an interpolation method. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s = pd.Series([10, None, 30, None, 50])
print("linear interpolation:", s.interpolate().tolist())
print("remaining missing:", s.interpolate().isna().sum())
print("✓ Example executed successfully; inspect the output against the question.")
```

**Explain `combine_first() / update() / align()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** combine_first fills missing values from another object; update overwrites matching non-missing values; align explicitly aligns labels. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
a = pd.Series([10, pd.NA, 30], index=["A","B","C"], dtype="Int64")
b = pd.Series([100, 200, 300], index=["A","B","C"], dtype="Int64")
print("combine_first:", a.combine_first(b).tolist())
c = a.copy()
c.update(b)
print("update:", c.tolist())
left, right = a.align(b)
print("aligned indexes equal:", left.index.equals(right.index))
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 70 — Module 70 — merge_asof() / merge_ordered()

---

## Question 1

**Explain `merge_asof() / merge_ordered()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** merge_asof matches ordered keys to the nearest eligible observation; merge_ordered combines ordered datasets while preserving order. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
left = pd.DataFrame({"time": pd.to_datetime(["2026-01-01 10:00","2026-01-01 11:00","2026-01-01 12:00"]), "price":[100,105,110]}).sort_values("time")
right = pd.DataFrame({"time": pd.to_datetime(["2026-01-01 10:30","2026-01-01 11:30"]), "event":["A","B"]}).sort_values("time")
print("asof:\n", pd.merge_asof(right, left, on="time"))
print("ordered:\n", pd.merge_ordered(left, right, on="time"))
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.
2. `left = pd.DataFrame({"time": pd.to_datetime(["2026-01-01 10:00","2026-01-01 11:00","2026-01-01 12:00"]), "price":[100,105,110]}).sort_values("time")` — `left = pd.DataFrame({"time": pd.to_datetime(["2026-01-01 10:00","2026-01-01 11:00","2026-01-01 12:00"]), "price":[100,105,110]}).sort_values("time")` specified column/value ke basis par rows ka order change karta hai.
3. `right = pd.DataFrame({"time": pd.to_datetime(["2026-01-01 10:30","2026-01-01 11:30"]), "event":["A","B"]}).sort_values("time")` — `right = pd.DataFrame({"time": pd.to_datetime(["2026-01-01 10:30","2026-01-01 11:30"]), "event":["A","B"]}).sort_values("time")` specified column/value ke basis par rows ka order change karta hai.
4. `print("asof:\n", pd.merge_asof(right, left, on="time"))` — `print("asof:\n", pd.merge_asof(right, left, on="time"))` relevant result/validation ko notebook output mein display karta hai.
5. `print("ordered:\n", pd.merge_ordered(left, right, on="time"))` — `print("ordered:\n", pd.merge_ordered(left, right, on="time"))` relevant result/validation ko notebook output mein display karta hai.
6. `print("✓ Example executed successfully; inspect the output against the question.")` — `print("✓ Example executed successfully; inspect the output against the question.")` relevant result/validation ko notebook output mein display karta hai.

# Module 71 — Module 71 — factorize()

---

## Question 1

**Explain `factorize()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** factorize converts repeated categories into integer codes plus the unique labels used for those codes. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s = pd.Series(["Gold","Silver","Gold","Bronze","Silver"])
codes, uniques = pd.factorize(s)
print("codes:", codes.tolist())
print("uniques:", uniques.tolist())
print("reconstructed:", uniques.take(codes).tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.

3. `codes, uniques = pd.factorize(s)` — `codes, uniques = pd.factorize(s)` repeated labels ko integer codes aur unique labels mein encode karta hai.
4. `print("codes:", codes.tolist())` — `print("codes:", codes.tolist())` relevant result/validation ko notebook output mein display karta hai.
5. `print("uniques:", uniques.tolist())` — `print("uniques:", uniques.tolist())` relevant result/validation ko notebook output mein display karta hai.

# Module 72 — Module 72 — GroupBy first() / last() / nth()

---

## Question 1

**Explain `GroupBy first() / last() / nth()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** GroupBy first, last, and nth select specific observations within each group. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"city":["K","K","K","L","L"],"sales":[100,120,150,80,90]})
g = df.groupby("city")["sales"]
print("first:\n", g.first())
print("last:\n", g.last())
print("nth(1):\n", g.nth(1))
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 74 — Module 74 — str.cat() / str.pad() / zfill()

---

## Question 1

**Explain `str.cat() / str.pad() / zfill()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** str.cat concatenates strings; str.pad adds padding; zfill pads numeric-looking strings with zeros. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
s = pd.Series(["Ali","Sara","Hamza"])
print("cat:", s.str.cat(sep=" | "))
print("padded:", s.str.pad(8, side="right", fillchar=".").tolist())
print("zfill:", pd.Series(["1","25","300"]).str.zfill(4).tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 75 — Module 75 — to_dict() / to_records()

---

## Question 1

**Explain `to_dict() / to_records()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** to_dict converts a DataFrame into Python dictionary structures; to_records converts rows into record-like structured data. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"name":["Ali","Sara"],"score":[80,90]})
print("dict records:", df.to_dict(orient="records"))
print("records:", df.to_records(index=False).tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 76 — Module 76 — iterrows() / itertuples()

---

## Question 1

**Explain `iterrows() / itertuples()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** iterrows iterates through rows as Series; itertuples iterates as tuples and is generally lighter for row-wise inspection. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"name":["Ali","Sara"],"score":[80,90]})
print("iterrows:")
for index, row in df.iterrows():
    print(index, row["name"], row["score"])
print("itertuples:")
for row in df.itertuples(index=False):
    print(row.name, row.score)
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 77 — Module 77 — tz_localize() / tz_convert() / Period

---

## Question 1

**Explain `tz_localize() / tz_convert() / Period` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** tz_localize attaches a timezone to naive timestamps; tz_convert changes an aware timestamp to another timezone; Period represents a calendar period. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
naive = pd.DatetimeIndex(["2026-01-01 12:00","2026-01-01 18:00"])
utc = naive.tz_localize("UTC")
karachi = utc.tz_convert("Asia/Karachi")
periods = pd.Series(pd.period_range("2026-01", periods=3, freq="M"))
print("UTC:", utc)
print("Karachi:", karachi)
print("Periods:", periods.tolist())
print("✓ Example executed successfully; inspect the output against the question.")
```

# Module 80 — Module 80 — wide_to_long()

---

## Question 1

**Explain `wide_to_long()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** wide_to_long reshapes repeated wide column patterns into a long format using a stub naming pattern. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
df = pd.DataFrame({"athlete":["Ali","Sara"],"score_2025":[80,90],"score_2026":[85,95]})
long = pd.wide_to_long(df, stubnames="score", i="athlete", j="year", sep="_", suffix="\d+")
print(long.reset_index())
print("✓ Example executed successfully; inspect the output against the question.")
```

### 🔎 Line-by-Line Explanation

1. `import pandas as pd` — Pandas ko `pd` short name ke saath import karta hai, taa-ke DataFrame/Series aur Pandas functions use kar saken.

# Module 81 — Module 81 — pd.testing.assert_frame_equal()

---

## Question 1

**Explain `pd.testing.assert_frame_equal()` in your own words and state the problem it solves. ### 💻 Coding Practice**

### ✅ Answer

**Answer:** assert_frame_equal verifies that two DataFrames match in values, labels, dtypes, and related structural details. The practical rule is to understand the input structure and expected output before executing the operation.

```python
import pandas as pd
expected = pd.DataFrame({"score":[90,95]}, index=["A","B"])
actual = pd.DataFrame({"score":[90,95]}, index=["A","B"])
pd.testing.assert_frame_equal(expected, actual)
print("Assertion passed: the two DataFrames are equivalent.")
print("✓ Example executed successfully; inspect the output against the question.")
```
