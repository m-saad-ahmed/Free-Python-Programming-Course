# Section 3 — Practice: Type Conversion

### Question 1 — Ask the user for a number, convert it to an integer, and print it.

```python
age_text = input("Enter your age: ")
age = int(age_text)

print(age)
```

**Line-by-line explanation**
1. `input(...)` always returns text (a String), even if the user types digits — so `age_text` holds something like `"25"`, not the number `25`.
2. `int(age_text)` converts that text into a real Integer.
3. `print(age)` displays the number, now usable in math (e.g. `age + 1` would work; `age_text + 1` would not).

### Question 2 — Ask for two integers and print their sum.

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

print(num1 + num2)
```

**Line-by-line explanation**
1. `int(input(...))` converts the user's text input into an Integer in one step, combining Question 1's two lines into one.
2. This is done twice, once for each number.
3. `num1 + num2` adds the two Integers together, and `print(...)` displays the sum.

### Question 3 — Ask for a decimal number, convert it to `float`, and print it.

```python
weight_text = input("Enter your weight: ")
weight = float(weight_text)

print(weight)
```

**Line-by-line explanation**
1. `input(...)` returns text, e.g. `"60.5"`.
2. `float(weight_text)` converts that text into a real Float, `60.5`.
3. `print(weight)` displays the number.

### Question 4 — Ask for a number and print its type after converting it to `int`.

```python
number = int(input("Enter a whole number: "))
print(type(number))
```

**Line-by-line explanation**
1. `int(input(...))` reads the input and immediately converts it to an Integer.
2. `type(number)` confirms the conversion worked by displaying `<class 'int'>`.

### Question 5 — Ask for a number as input, convert it to an integer, and calculate its double.

```python
number = int(input("Enter a number: "))
double = number * 2

print(double)
```

**Line-by-line explanation**

# Section 6 — Practice: Boolean Values / Logical Operators

### Question 1 — Create two Boolean variables and combine them with `and`.

```python
has_ticket = True
has_id = True

print(has_ticket and has_id)
```

**Line-by-line explanation**
1. `has_ticket` and `has_id` are both `True`.
2. `and` requires **both** to be `True` to give `True` — since both are `True`, the result is `True`.

### Question 2 — Create two Boolean variables and combine them with `or`.

```python
has_cash = False
has_card = True

print(has_cash or has_card)
```

**Line-by-line explanation**
1. `has_cash` is `False`, `has_card` is `True`.
2. `or` only needs **one** side to be `True` — since `has_card` is `True`, the result is `True`.

### Question 3 — Create a Boolean variable and reverse it using `not`.

```python
is_raining = False
print(not is_raining)
```

**Line-by-line explanation**
1. `is_raining` is `False`.
2. `not` reverses a Boolean, so `not False` becomes `True`.

### Question 4 — Create an age condition and an ID condition and combine them with `and`.

```python
age = 20
has_id = True

can_enter = (age >= 18) and has_id
print(can_enter)
```

**Line-by-line explanation**
1. `age >= 18` evaluates to `True` (since `age` is `20`).
2. `has_id` is `True`.
3. `and` combines both conditions — since both are `True`, `can_enter` becomes `True`. This pattern (combining a comparison with a Boolean using `and`) is exactly how real-world access-control logic is written.

### Question 5 — Create two conditions and build one expression using both `and` and `or`.

```python
age = 16
has_guardian = True
has_special_pass = False

can_watch = (age >= 18) or (has_guardian and age >= 13)
print(can_watch)
```

**Line-by-line explanation**
1. `age >= 18` → `False` (age is `16`).
2. `has_guardian and age >= 13` → `has_guardian` is `True` **and** `age >= 13` is `True`, so this combined expression is `True`.
3. The full expression is `False or True`, and since `or` only needs one side `True`, `can_watch` becomes `True`.
4. This shows how `and`/`or` can be combined in one line to express a real rule: "allowed if 18+, OR if accompanied by a guardian and at least 13."
