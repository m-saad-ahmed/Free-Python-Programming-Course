### Question 1
Create a test repository on GitHub.

**Solution (steps, not code — this happens on the GitHub website):**
1. Go to [github.com](https://github.com) and log in.
2. Click the **New repository** button.
3. Give it a name, e.g. `test-repo`.
4. Leave it **Public** or **Private** (either works for practice).
5. Click **Create repository**.

**Explanation**
- This step doesn't involve writing any code — it happens entirely in your browser, on GitHub's website.
- GitHub then gives you a repository **URL**, something like `https://github.com/<your-username>/test-repo.git`. You'll need this exact URL later, in Question 7.
- Creating the repository *first* (before touching the terminal) is the standard order, because the remaining steps connect your local project *to* this repository — it needs to already exist.

---

### Question 2
Create a small project folder on your computer.

```python
mkdir test-repo
cd test-repo
```

**Line-by-line explanation**
- `mkdir test-repo` → `mkdir` (make directory) creates a new, empty folder named `test-repo` on your computer. Naming it the same as your GitHub repository (from Question 1) is a helpful convention, not a requirement.
- `cd test-repo` → `cd` (change directory) moves your terminal *into* that new folder, so any commands you run next apply to this project, not somewhere else on your computer.

---

### Question 3
Open the project folder in the terminal.

```python
cd path/to/test-repo
```

**Line-by-line explanation**
- `cd path/to/test-repo` → if you're not already inside the folder (for example, if you closed and reopened your terminal), `cd` followed by the folder's path takes you back into it.
- Replace `path/to/test-repo` with the actual location on your computer, e.g. `cd Desktop/test-repo` or `cd /Users/yourname/test-repo`.
- Every Git command from this point on must be run **from inside** this project folder — Git commands act on whatever folder you are currently in.

---

### Question 4
Run `git init`.

```python
git init
```

**Line-by-line explanation**
- `git init` → tells Git: "start tracking this folder as a Git project."
- This creates a hidden subfolder called `.git` inside your project. That hidden folder is where Git will quietly store the entire history of every change you make — you never touch it directly.
- After this command, your plain folder officially becomes a **Git repository**.

---

### Question 5
Run `git add .`.

```python
git add .
```

**Line-by-line explanation**
- `git add .` → stages files, meaning it tells Git: "these are the files I want to include in my next checkpoint (commit)."
- The `.` means "everything in the current folder" — every new or changed file gets staged at once.
- Staging is a separate step from committing on purpose — it lets you review and select exactly what goes into a commit, rather than committing every file automatically.

---

### Question 6
Create your first commit.

```python
git commit -m "first commit"
```

**Line-by-line explanation**
- `git commit` → takes everything that was staged in Question 5 and saves it as a permanent **checkpoint** in your project's history.
- `-m "first commit"` → the `-m` flag lets you attach a short message describing what this checkpoint contains. Every commit needs a message; `"first commit"` is the conventional message for the very first one.
- After this line runs, Git has now permanently recorded a snapshot of your project that you can always come back to.

---

### Question 7
Connect the project to the GitHub repository.

```python
git remote add origin https://github.com/<your-username>/test-repo.git
```

**Line-by-line explanation**
- `git remote add` → tells Git: "I want to link this local project to an online (remote) repository."
- `origin` → this is just a **nickname** for the remote repository's address. `origin` is the standard, conventional name almost everyone uses for their main remote.
- `https://github.com/<your-username>/test-repo.git` → this is the actual repository URL — the one GitHub gave you back in Question 1. You must replace `<your-username>` with your real GitHub username (and `test-repo` with your real repository name, if different).
- ⚠️ As the class notes warn: **never copy someone else's repository URL** — always use the URL of the repository *you* created.

---

### Question 8
Push the project to GitHub.

```python
git branch -M main
git push -u origin main
```

**Line-by-line explanation**
- `git branch -M main` → renames your current branch to `main`. Modern GitHub expects your primary branch to be named `main` (older Git versions sometimes default to `master`), so this line makes sure the naming matches.
- `git push -u origin main` → `push` uploads your committed project from your computer to GitHub.
  - `origin` → the remote repository we linked in Question 7.
  - `main` → the branch being uploaded.
  - `-u` → sets this as the **default** link between your local `main` branch and GitHub's `main` branch, so that in the future you can simply type `git push` without repeating `origin main` every time.
- After this runs successfully, refreshing your GitHub repository page in the browser will show your uploaded files.

## Quick Reference — The Full Sequence Together

For convenience, here is the entire 8-step sequence as one block, exactly as you'd type it for a real project:

```python
mkdir test-repo
cd test-repo

git init
git add .
git commit -m "first commit"

git remote add origin https://github.com/<your-username>/test-repo.git
git branch -M main
git push -u origin main
```

**Explanation**
- The first two lines set up the local folder.
- The next three lines turn it into a Git project and save your first checkpoint.
- The final three lines connect that project to GitHub and upload it.
- Remember: `<your-username>` (and the repository name, if you chose something other than `test-repo`) must be replaced with your **own** real values every time — this is the one part of the sequence that is never copy-pasted as-is.
