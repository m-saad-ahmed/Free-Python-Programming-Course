## Practice — `zip()`

### Question 1
Zip three student names with three scores.

```python
names = ["Ali", "Ahmed", "Saad"]
scores = [85, 90, 78]

for name, score in zip(names, scores):
    print(name, score)
```

**Line-by-line explanation**
- `zip(names, scores)` → pairs up items from both lists **position by position**: the 1st name with the 1st score, the 2nd with the 2nd, and so on.
- `for name, score in zip(...):` → unpacks each pair into two variables.
- `print(name, score)` → prints each name next to its matching score.

### Question 2
Zip three cities with three temperatures.

```python
cities = ["Karachi", "Lahore", "Multan"]
temperatures = [34, 29, 38]

for city, temp in zip(cities, temperatures):
    print(city, temp)
```

**Line-by-line explanation**
- Same idea: `zip()` combines the two lists position by position, so `"Karachi"` is paired with `34`, `"Lahore"` with `29`, and so on.

### Question 3
Print each pair from two equal-length lists.

```python
letters = ["A", "B", "C"]
numbers = [1, 2, 3]

for letter, number in zip(letters, numbers):
    print(letter, number)
```

**Line-by-line explanation**
- Because both lists have exactly 3 items, `zip()` produces exactly 3 pairs, with nothing left unmatched.

### Question 4
Try two lists with different lengths and observe what happens.

```python
names = ["Ali", "Ahmed", "Saad", "Hina"]
scores = [85, 90]

for name, score in zip(names, scores):
    print(name, score)
```

**Line-by-line explanation**
- `names` has 4 items, but `scores` has only 2.
- `zip()` stops as soon as the **shorter** list runs out of items — so only 2 pairs are produced (`Ali`/`85` and `Ahmed`/`90`), and `"Saad"` and `"Hina"` are silently ignored because they have no matching score.
- This is an important behavior to know: `zip()` never raises an error for mismatched lengths, it just stops early.

### Question 5
Use `zip()` to print a name together with a corresponding subject.

```python
names = ["Ali", "Ahmed", "Saad"]
subjects = ["Math", "Physics", "Chemistry"]

for name, subject in zip(names, subjects):
    print(name, "studies", subject)
```

**Line-by-line explanation**
- `zip(names, subjects)` → pairs each name with the subject at the same position.
- `print(name, "studies", subject)` → combines them into a readable sentence, e.g. `Ali studies Math`.

## Practice — Nested loops

### Question 1
Create two lists of numbers and compare their items.

```python
list_a = [1, 2, 3]
list_b = [2, 3, 4]

for a in range(len(list_a)):
    for b in range(len(list_b)):
        if list_a[a] == list_b[b]:
            print(list_a[a], "matches", list_b[b])
```

**Line-by-line explanation**
- `for a in range(len(list_a)):` → the **outer loop** walks through every valid index of `list_a`.
- `for b in range(len(list_b)):` → for **each** value of `a`, the **inner loop** walks through every valid index of `list_b`. This is what makes it "nested" — the inner loop fully completes for every single step of the outer loop.
- `if list_a[a] == list_b[b]:` → compares the actual values (not the indexes) at those positions.
- The result prints every pair of matching numbers found between the two lists (here, `2` and `3`).

### Question 2
Create two lists of cities and print every possible pair.

```python
group_a = ["Karachi", "Lahore"]
group_b = ["Multan", "Quetta"]

for a in range(len(group_a)):
    for b in range(len(group_b)):
        print(group_a[a], group_b[b])
```

**Line-by-line explanation**
- Unlike Question 1, there's no `if` condition here — every combination is printed, regardless of matching.
- Because the outer loop runs 2 times and the inner loop runs 2 times *for each* outer step, this produces `2 × 2 = 4` printed pairs in total.

### Question 3
Use nested loops to compare two lists of names.

```python
names_a = ["Ali", "Ahmed", "Saad"]
names_b = ["Saad", "Hina"]

for a in range(len(names_a)):
    for b in range(len(names_b)):
        if names_a[a] == names_b[b]:
            print(names_a[a], "is in both lists")
```

**Line-by-line explanation**
- Same structure as Question 1, applied to names instead of numbers.
- The nested loop checks every name in `names_a` against every name in `names_b`, and the `if` only prints when it finds an exact match — here, `"Saad"`.

### Question 4
Add a condition that prints `Matched` when two values are equal.

```python
list_a = [5, 10, 15]
list_b = [10, 20, 15]

for a in range(len(list_a)):
    for b in range(len(list_b)):
        if list_a[a] == list_b[b]:
            print("Matched:", list_a[a])
```

**Line-by-line explanation**
- Identical logic to before, but this time the printed message explicitly says `"Matched:"` before the value, making the output's meaning clearer to read.

### Question 5
Use `break` when a match is found.

```python
list_a = [5, 10, 15]
list_b = [10, 20, 15]

for a in range(len(list_a)):
    for b in range(len(list_b)):
        if list_a[a] == list_b[b]:
            print("Matched:", list_a[a])
            break
```

**Line-by-line explanation**
- `break` → immediately stops the **innermost** loop it's placed in — here, the inner `b` loop.
- As soon as a match for the current `a` is found, there's no need to keep checking the rest of `list_b` for that same `a` — `break` exits early and moves the outer loop on to the next value of `a`.
- Note that `break` only exits the inner loop, not both loops — the outer `a` loop keeps running normally.

## Practice — `continue`

### Question 1
Print numbers 1 to 10, but skip the number 5 using `continue`.

```python
for number in range(1, 11):
    if number == 5:
        continue
    print(number)
```

**Line-by-line explanation**
- `for number in range(1, 11):` → loops through `1` to `10`.
- `if number == 5: continue` → when `number` is exactly `5`, `continue` immediately jumps back to the **top of the loop**, skipping everything below it for this iteration — meaning `print(number)` never runs when `number` is `5`.
- Every other number still gets printed normally.

### Question 2
Loop through a list of names and skip printing your own name.

```python
names = ["Ali", "Ahmed", "Saad", "Hina"]
my_name = "Saad"

for name in names:
    if name == my_name:
        continue
    print(name)
```

**Line-by-line explanation**
- `if name == my_name: continue` → whenever the current `name` matches `my_name`, the loop skips straight to the next name without executing `print(name)`.
- Every name except `"Saad"` gets printed.

### Question 3
Loop through numbers 1 to 20 and use `continue` to skip every multiple of 3.

```python
for number in range(1, 21):
    if number % 3 == 0:
        continue
    print(number)
```

**Line-by-line explanation**
- `number % 3 == 0` → the `%` (modulo) operator gives the remainder of dividing `number` by `3`. When the remainder is `0`, the number is a multiple of `3`.
- `continue` skips printing for every such multiple (`3, 6, 9, 12, 15, 18`), while every other number is printed normally.

### Question 4
Combine `continue` and `if` to skip negative numbers in a list.

```python
numbers = [4, -2, 7, -9, 10, -1]

for number in numbers:
    if number < 0:
        continue
    print(number)
```

**Line-by-line explanation**
- `if number < 0: continue` → whenever a number is negative, the loop skips it immediately.
- Only the non-negative numbers (`4, 7, 10`) end up being printed.

### Question 5
Explain in your own words the difference between `break` and `continue`.

**Answer (conceptual — no code needed)**

- `break` **completely stops** the loop — as soon as Python reaches `break`, it exits the loop entirely and moves on to whatever code comes after it. No further iterations happen at all.
- `continue` only **skips the current iteration** — Python jumps back to the top of the loop and moves on to the *next* item, but the loop itself keeps running normally.

**Simple analogy:** if a loop were a to-do list you're working through one item at a time, `break` means stop the whole list right now, while `continue` means skip just this one item and move on to the next.

## Practice — `while` loops

### Question 1
Print numbers from 1 to 5 using `while`.

```python
number = 1

while number <= 5:
    print(number)
    number = number + 1
```

**Line-by-line explanation**
- `number = 1` → starts a counter at `1`.
- `while number <= 5:` → the loop keeps running **as long as** this condition stays `True`.
- `print(number)` → prints the current counter value.
- `number = number + 1` → increases the counter by `1` each time. This step is essential — without it, `number` would stay `1` forever and the loop would never end (an infinite loop).

### Question 2
Print even numbers from 2 to 10 using `while`.

```python
number = 2

while number <= 10:
    print(number)
    number = number + 2
```

**Line-by-line explanation**
- Instead of incrementing by `1`, `number = number + 2` jumps straight to the next even number each time, starting from `2`.
- The loop naturally produces `2, 4, 6, 8, 10` and then stops once `number` exceeds `10`.

### Question 3
Calculate the sum from 1 to 5.

```python
number = 1
total = 0

while number <= 5:
    total = total + number
    number = number + 1

print("Total:", total)
```

**Line-by-line explanation**
- `total = 0` → an accumulator variable that starts at zero.
- Each time through the loop, `total = total + number` adds the current counter value onto the running total.
- After the loop finishes (once `number` becomes `6` and the condition fails), `total` holds `1 + 2 + 3 + 4 + 5 = 15`.

### Question 4
Print the multiplication table of 5.

```python
num = 5
i = 1

while i <= 10:
    result = num * i
    print(num, "x", i, "=", result)
    i = i + 1
```

**Line-by-line explanation**
- `num = 5` → the number whose table we want.
- `i = 1` → the multiplier, starting at `1`.
- `while i <= 10:` → repeats for multipliers `1` through `10`.
- `result = num * i` → calculates each line of the table.
- `i = i + 1` → moves to the next multiplier so the loop eventually ends.

### Question 5
Create a counter that stops after 10.

```python
counter = 1

while counter <= 10:
    print("Counter is at:", counter)
    counter = counter + 1

print("Done!")
```

**Line-by-line explanation**
- `counter = 1` → starts the counter.
- The loop prints the counter's current value and increases it by `1` every time, until `counter` exceeds `10`, at which point the condition becomes `False` and the loop stops.
- `print("Done!")` runs once, after the loop has fully finished.

## Practice — Country checking

### Question 1
Create a list of five countries.

```python
countries = ["Pakistan", "Japan", "Germany", "Brazil", "Canada"]
print(countries)
```

**Line-by-line explanation**
- `countries = [...]` → a plain list of five country name strings, which the rest of this section will check against.

### Question 2
Ask the user for a country.

```python
user_input = input("Enter a country: ")
print(user_input)
```

**Line-by-line explanation**
- `input("Enter a country: ")` → pauses the program, shows the message `"Enter a country: "`, and waits for the user to type something and press Enter.
- Whatever the user types is returned as a **String** and stored in `user_input`.
- ⚠️ Because this uses `input()`, running this cell yourself will actually pause and wait for you to type a value.

### Question 3
Use a `for` loop to check the list.

```python
countries = ["Pakistan", "Japan", "Germany", "Brazil", "Canada"]
user_input = input("Enter a country: ")

for country in countries:
    if country == user_input:
        print("Found it!")
```

**Line-by-line explanation**
- `for country in countries:` → checks every country in the list, one at a time.
- `if country == user_input:` → compares each one to what the user typed. If it's an exact match, `"Found it!"` is printed.
- Note: this simple version keeps checking every remaining country even after finding a match — that's what Question 4 improves on.

### Question 4
Use `break` after a match.

```python
countries = ["Pakistan", "Japan", "Germany", "Brazil", "Canada"]
user_input = input("Enter a country: ")

for country in countries:
    if country == user_input:
        print("Found it!")
        break
```

**Line-by-line explanation**
- `break` → as soon as a match is found and `"Found it!"` is printed, the loop stops checking immediately, instead of wastefully comparing against the remaining countries in the list.
- This makes the search more efficient without changing the result.

### Question 5
Create a `while` version that keeps asking until the user enters a valid country.

```python
countries = ["Pakistan", "Japan", "Germany", "Brazil", "Canada"]

while True:
    user_input = input("Enter a valid country: ")
    if user_input in countries:
        print("Thank you, valid country entered!")
        break
    else:
        print("Not in the list, try again.")
```

**Line-by-line explanation**
- `while True:` → creates a loop that would normally run forever, since `True` is always `True` — this pattern is only safe because there's a `break` inside to stop it.
- `if user_input in countries:` → the `in` keyword checks membership directly, without needing a separate `for` loop.
- If the input matches, the program prints a success message and `break`s out of the `while True:` loop.
- If not, the `else` branch prints an error message, and since there's no `break` there, the loop repeats and asks again.

## Practice — More string methods

### Question 1
Use `.strip()` to clean up a string with extra spaces around it.

```python
text = "   Hello World   "
print(text.strip())
```

**Line-by-line explanation**
- `.strip()` → removes any leading and trailing whitespace (spaces, tabs, newlines) from a string, but leaves spaces **inside** the text untouched.
- `"   Hello World   "` becomes `"Hello World"`.

### Question 2
Use `.split()` to break a sentence into a list of words.

```python
sentence = "Python is a powerful language"
words = sentence.split()
print(words)
```

**Line-by-line explanation**
- `.split()` → with no argument, splits the string wherever there's whitespace, and returns a **list** of the individual words.
- The result is `['Python', 'is', 'a', 'powerful', 'language']`.

### Question 3
Use `.replace()` to change one word in a sentence to another word.

```python
sentence = "I love JavaScript"
new_sentence = sentence.replace("JavaScript", "Python")
print(new_sentence)
```

**Line-by-line explanation**
- `.replace(old, new)` → returns a new string with every occurrence of `old` swapped out for `new`.
- Here, every occurrence of `"JavaScript"` is replaced with `"Python"`, producing `"I love Python"`. The original `sentence` variable is not changed.

### Question 4
Use `.find()` to locate a word inside a sentence.

```python
sentence = "Python is a powerful language"
position = sentence.find("powerful")
print(position)
```

**Line-by-line explanation**
- `.find(substring)` → returns the **index** where the substring first begins inside the string, or `-1` if it isn't found at all.
- `"powerful"` starts at character position `12` in this sentence, so `position` becomes `12`.

Combine `.strip()` and `.split()` in a single line, like the example above.

**Line-by-line explanation**
- `sentence.strip()` runs first, removing the extra spaces from the start and end.
- `.split(", ")` is then called directly on that result (this is called **method chaining**), splitting the cleaned string wherever it sees a comma followed by a space.
- The final result is a clean list: `['Karachi', 'Lahore', 'Islamabad']`.

## Practice — Dictionary loops

### Question 1
Print all keys.

```python
customer = {
    "first name": "Saad",
    "last name": "Ahmed",
    "city": "Karachi"
}

for key in customer.keys():
    print(key)
```

**Line-by-line explanation**
- `.keys()` → returns a view of all the keys in the dictionary.
- `for key in customer.keys():` → loops through them one at a time, printing each key on its own line.
- Note: you could also write `for key in customer:` directly, without `.keys()` — looping over a dictionary gives you its keys by default. Using `.keys()` explicitly just makes the intent extra clear.

### Question 2
Print all values.

```python
customer = {
    "first name": "Saad",
    "last name": "Ahmed",
    "city": "Karachi"
}

for value in customer.values():
    print(value)
```

**Line-by-line explanation**
- `.values()` → returns a view of all the **values** in the dictionary, ignoring the keys entirely.
- The loop prints each value (`"Saad"`, `"Ahmed"`, `"Karachi"`) on its own line.

### Question 3
Print every key with its value.

```python
customer = {
    "first name": "Saad",
    "last name": "Ahmed",
    "city": "Karachi"
}

for key, value in customer.items():
    print(key, "-->", value)
```

**Line-by-line explanation**
- `.items()` → returns each key-value pair together, as a tuple.
- `for key, value in customer.items():` → unpacks each pair directly into two variables in one step.
- `print(key, "-->", value)` → shows both together clearly, e.g. `first name --> Saad`.

### Question 4
Create a student dictionary and loop through `.items()`.

```python
student = {
    "name": "Ali",
    "age": 20,
    "grade": "A"
}

for key, value in student.items():
    print(key, "-->", value)
```

**Line-by-line explanation**
- Same `.items()` pattern as Question 3, applied to a new dictionary — showing that this loop style works for any dictionary, regardless of what it stores.

### Question 5
Create a product dictionary and print each key-value pair.

```python
product = {
    "name": "Laptop",
    "price": 150000,
    "brand": "Dell"
}

for key, value in product.items():
    print(key, "-->", value)
```

**Line-by-line explanation**
- One more `.items()` example, confirming this is the standard, professional way to loop through both a dictionary's keys and values together whenever you need both.

## Practice — Lists of dictionaries

### Question 1
Create a list containing three student dictionaries.

```python
students = [
    {"name": "Ali", "city": "Karachi"},
    {"name": "Ahmed", "city": "Lahore"},
    {"name": "Saad", "city": "Islamabad"}
]

print(students)
```

**Line-by-line explanation**
- `students = [...]` → a **list** where each individual item is itself a **dictionary**. This combination is extremely common in real-world data (e.g. a list of user records).

### Question 2
Print the first student's name.

```python
students = [
    {"name": "Ali", "city": "Karachi"},
    {"name": "Ahmed", "city": "Lahore"},
    {"name": "Saad", "city": "Islamabad"}
]

print(students[0]["name"])
```

**Line-by-line explanation**
- `students[0]` → uses the **list's index** to get the first dictionary (`{"name": "Ali", "city": "Karachi"}`).
- `["name"]` → then uses the **dictionary's key** on that result to get its `"name"` value.
- Reading it left to right: "get item 0 from the list, then get the 'name' key from that item." This prints `"Ali"`.

### Question 3
Print the second student's city.

```python
students = [
    {"name": "Ali", "city": "Karachi"},
    {"name": "Ahmed", "city": "Lahore"},
    {"name": "Saad", "city": "Islamabad"}
]

print(students[1]["city"])
```

**Line-by-line explanation**
- `students[1]` → the second dictionary in the list (index `1`, `Ahmed`'s record).
- `["city"]` → looks up the `"city"` key inside that dictionary, printing `"Lahore"`.

### Question 4
Loop through the list and print every student's name.

```python
students = [
    {"name": "Ali", "city": "Karachi"},
    {"name": "Ahmed", "city": "Lahore"},
    {"name": "Saad", "city": "Islamabad"}
]

for student in students:
    print(student["name"])
```

**Line-by-line explanation**
- `for student in students:` → each pass through the loop, `student` becomes one full dictionary from the list (not just a name).
- `student["name"]` → inside the loop, we pull out just the `"name"` key from whichever dictionary we currently have.
- This prints every student's name, one per line, without needing to know how many students there are in advance.

### Question 5
Loop through the list and print every student's city.

```python
students = [
    {"name": "Ali", "city": "Karachi"},
    {"name": "Ahmed", "city": "Lahore"},
    {"name": "Saad", "city": "Islamabad"}
]

for student in students:
    print(student["city"])
```

**Line-by-line explanation**
- Identical loop structure to Question 4, just reading the `"city"` key instead of `"name"` — showing how easily the same loop pattern adapts to pull out any field you need from each dictionary.
